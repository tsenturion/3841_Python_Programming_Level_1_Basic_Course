### 3.4 **Практическое задание «Поиск числа»**

Напишите программу, которая перебирает числа от 1 до 20. Пользователь вводит число, которое необходимо найти.

Программа должна:

- Пропускать все чётные числа с помощью `continue`.
- При обнаружении искомого числа выводить сообщение «Число найдено» и завершать цикл с помощью `break`.
- Если число не найдено, выводить сообщение «Число не найдено» с помощью блока `else` цикла.

**Пример работы программы:**

```
Введите число: 7
Число найдено
```

```
Введите число: 8
Число не найдено
```

**Ограничения:** использовать цикл `for`, операторы `break`, `continue` и блок `else` цикла.

````python
search_number = int(input("Введите число: "))

for number in range(1, 21):
  if number % 2 == 0:
    continue

  if number == search_number:
    print("Число найдено")
    break
else:
  print("Число не найдено")
````

