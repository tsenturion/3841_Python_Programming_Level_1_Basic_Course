## 1.1 Установка интерпретатора Python и IDE;

**Пример 1. Две команды вывода**

Этой программой можно проверить, что выбранный интерпретатор выполняет Python-код, а среда разработки показывает стандартный вывод. В Jupyter напечатанный текст появляется под ячейкой. При запуске файла из терминала он появляется в терминале.

Каждая пара кавычек ограничивает строковый литерал. Python передаёт строку между кавычками функции `print()`. По умолчанию `print()` завершает вывод переводом строки, поэтому два вызова дают две строки в порядке выполнения.

Измените только текст внутри второй пары кавычек и снова запустите ячейку. Первое сообщение останется прежним, а второе изменится. Для строки используйте совпадающие открывающую и закрывающую кавычки, иначе Python сообщит о синтаксической ошибке ещё при разборе ячейки.

**Ожидаемый вывод**

```text
Учебный помощник
Программа запущена успешно
```

````python
# Вызов print получает строку с названием программы. После печати функция переводит
# строку.
print("Учебный помощник")
# Второй вызов печатает сообщение о запуске на следующей строке.
print("Программа запущена успешно")
````

**Пример 2. Приветствие из трёх строк**

Три последовательных вызова `print()` образуют приветствие. Первый сообщает название программы, второй обращается к слушателю, третий показывает текст о готовности рабочей среды.

Интерпретатор выполняет вызовы сверху вниз. Порядок сообщений зависит от положения команд в ячейке. Кавычки в исходнике помогают Python найти границы строки и при выводе отбрасываются. Кириллица и восклицательный знак внутри строки сохраняются.

Каждый запуск выполняет все три команды заново. Фраза «Рабочая среда настроена» задаётся литералом, поэтому её появление подтверждает выполнение команды печати. Для проверки конкретного свойства среды, например версии Python, нужна отдельная проверка.

**Ожидаемый вывод**

```text
Учебный помощник
Добро пожаловать на курс Python!
Рабочая среда настроена
```

````python
# Напечатаем название учебного помощника первой строкой.
print("Учебный помощник")
# Добавим отдельную строку с приветствием. Восклицательный знак находится внутри
# строкового литерала и попадёт в вывод.
print("Добро пожаловать на курс Python!")
# Завершим приветствие текстом о готовности среды. Это заранее заданное сообщение.
print("Рабочая среда настроена")
````

## 1.2 Неизменяемые типы данных: int, float, str, bool;

**Пример 3. Имя для строки и имя для целого числа**

Оператор `=` связывает имя слева со значением справа. Имя `course_name` указывает на строку `str`, а `students_count` указывает на целое число `int`. Тип определяется записанным значением, поскольку объявлять тип имени для такого присваивания в Python не требуется.

В обеих командах `print()` без кавычек указаны имена. Python находит текущее значение каждого имени и печатает его. Значение `18` выводится без дополнительной подписи, поскольку у второго вызова один аргумент.

Вариант `print("course_name")` напечатал бы буквальный текст `course_name`. Имена с подчёркиваниями читаются как несколько слов. Пустая строка отделяет подготовку данных от их вывода и служит для читаемости.

**Ожидаемый вывод**

```text
Программирование на Python
18
```

````python
# Связываем имя course_name со строкой. Кавычки ограничивают текст значения.
course_name = "Программирование на Python"
# Сохраняем количество слушателей как целое число 18. В литерале отсутствуют кавычки и
# десятичная точка.
students_count = 18

# По имени course_name Python получает строку с полным названием курса и передаёт её
# функции print.
print(course_name)
# Печатаем числовое значение students_count. Оно оказалось на новой строке после первого
# вызова print.
print(students_count)
````

**Пример 4. Целые числа и проверка типа**

Значения `18`, `20` и `0` — целочисленные литералы. В учебном примере ими удобно представлять счётчики студентов, занятий и свободных мест. Ноль здесь обозначает фактическое количество мест.

Первые три вызова `print()` показывают значения. В последней строке сначала вычисляется внутренний вызов `type(students_count)`. Он даёт объект класса `int`. Внешний `print()` печатает его представление `<class 'int'>`.

Текст `"18"` имел бы тип `str`, а `18.0` — тип `float`. Внешне числа могут выглядеть похоже, поэтому при разборе неожиданного поведения `type()` помогает проверить фактический тип.

**Ожидаемый вывод**

```text
18
20
0
<class 'int'>
```

````python
# Сохраняем количество студентов целочисленным литералом 18.
students_count = 18
# Сохраняем запланированное количество занятий целочисленным литералом 20.
lessons_count = 20
# Указываем, что свободных мест ноль. У 0 тот же числовой тип int.
available_seats = 0

# Выводим число студентов. В вывод попадает значение 18.
print(students_count)
# Выводим число занятий. Следующая строка содержит значение 20.
print(lessons_count)
# Выводим количество свободных мест, включая значение 0.
print(available_seats)

# Внутренний type возвращает класс int для числа студентов, внешний print показывает этот
# класс.
print(type(students_count))
````

**Пример 5. Дробные величины и тип float**

Числовые литералы `1.5`, `4.8` и `0.15` содержат десятичную точку и создают значения типа `float`. Продолжительность `1.5` означает полтора часа. Скидка `0.15` хранится как доля единицы и соответствует 15 процентам.

Три команды печати показывают значения по одному. Проверка `type(lesson_duration)` демонстрирует, что продолжительность хранится как `float`. Рейтинг и скидка заданы тем же синтаксисом и имеют тот же тип.

Для десятичной дроби в исходном коде нужна точка. Например, `discount = 0, 15` создаёт кортеж из двух целых чисел. Формат вывода можно выбирать отдельно от способа хранения значения.

**Ожидаемый вывод**

```text
1.5
4.8
0.15
<class 'float'>
```

````python
# Записываем полтора часа числом 1.5. Наличие точки задаёт тип float.
lesson_duration = 1.5
# Задаём оценку курса 4.8 как дробное число.
course_rating = 4.8
# Задаём скидку долей единицы. Число 0.15 представляет 15 процентов.
discount = 0.15

# Печатаем продолжительность занятия в часах.
print(lesson_duration)
# Печатаем дробное значение рейтинга курса.
print(course_rating)
# Печатаем скидку в исходном виде, то есть десятичную долю.
print(discount)

# Через type получаем класс значения lesson_duration и выводим <class 'float'>.
print(type(lesson_duration))
````

**Пример 6. Почему сумма 0.1 и 0.2 выглядит длиннее ожидаемого**

Первые четыре строки вывода повторяют знакомство с `float`. Последний вызов оценивает, что произойдёт при сложении двух десятичных дробей, представленных в двоичном формате.

В обычном Python значения `0.1` и `0.2` хранятся как ближайшие представимые двоичные числа. При сложении результат снова округляется до представимого `float`. Поэтому обычный вывод показывает `0.30000000000000004`. Число `1.5` из того же примера имеет точное двоичное разложение `1 + 1/2`.

При сравнении измерений полезно проверять близость через `math.isclose()` с требуемым допуском. Если расчёт требует контролируемой десятичной точности, например в финансовых задачах, можно начать с `Decimal('0.1')` из модуля `decimal`.

**Ожидаемый вывод**

```text
1.5
4.8
0.15
<class 'float'>
0.30000000000000004
```

````python
# Сохраняем значение 1.5. Половина представима в двоичном формате точно.
lesson_duration = 1.5
# Сохраняем рейтинг 4.8 как float. Двоичное представление этого значения приближённое.
course_rating = 4.8
# Сохраняем скидку 0.15 как float. При выводе используется короткая десятичная запись.
discount = 0.15

# Печатаем длительность занятия для сравнения с остальными дробными значениями.
print(lesson_duration)
# Выводим рейтинг в обычном представлении float.
print(course_rating)
# Выводим скидку как долю единицы.
print(discount)

# Подтверждаем через type, что длительность имеет тип float.
print(type(lesson_duration))
# Сначала складываем два float, затем печатаем результат с видимым эффектом двоичного
# округления.
print(0.1 + 0.2)
````

**Пример 7. Хранение и вывод текста**

Три значения заключены в двойные кавычки и имеют тип `str`. Название содержит пробелы и кириллицу, имя состоит из нескольких букв, а в коде курса есть буквы, дефис и цифры. Все эти символы остаются содержимым строки.

`print()` последовательно выводит название, имя и код. Кавычки ограничивают литералы в программе и в эти строки вывода не входят. Последний вызов проверяет тип названия и показывает `<class 'str'>`.

Код курса `"PY-101"` хранится как строка, поэтому дефис воспринимается как обычный символ. Текст в таком формате удобно использовать как идентификатор, в том числе если у цифровой части появятся ведущие нули.

**Пример 8. Одинарные и двойные кавычки**

Строковый литерал в Python можно ограничить одинаковыми одинарными или одинаковыми двойными кавычками. Значения `'Александр'` и `"Александр"` содержат один и тот же текст и имеют тип `str`.

В этой версии имя записано в одинарных кавычках, а два других значения — в двойных. После разбора каждый вызов `print()` получает уже готовую строку, поэтому изменение внешних кавычек у имени никак не отражается на напечатанных символах.

Хороший способ проверить правило — вычислить выражение `'Александр' == "Александр"`. Оно даст `True`. Выбор внешних кавычек становится заметен, когда внутри текста встречается кавычка того же вида.

**Ожидаемый вывод**

```text
Программирование на Python
Александр
PY-101
<class 'str'>
```

````python
# Записываем название курса строкой с двойными кавычками.
course_name = "Программирование на Python"
# Записываем имя строкой с одинарными кавычками. Сами ограничители не входят в значение.
teacher_name = 'Александр'
# Оставляем код курса в двойных кавычках для сравнения двух форм записи.
course_code = "PY-101"

# Печатаем название курса на первой строке.
print(course_name)
# Печатаем имя. В выводе появятся только буквы имени.
print(teacher_name)
# Печатаем код курса после имени.
print(course_code)

# Показываем класс названия курса. Для обоих видов кавычек создаётся строка типа str.
print(type(course_name))
````

**Пример 9. Кавычка как часть строкового значения**

Внешний ограничитель задаёт, по какой неэкранированной кавычке закончится литерал. В строке `'Алекс"андр'` внешние одинарные кавычки позволяют оставить двойную кавычку между буквами. В строке `"PY-1'01"` двойные кавычки позволяют включить апостроф.

При выводе в строке имени видно `Алекс"андр`, а в строке кода — `PY-1'01`. В обоих случаях печатается внутренняя кавычка. Вывод класса названия курса показывает, что все записи такого вида представляют строки.

Для кавычки того же вида, что и внешняя, можно использовать обратную косую черту. Например, `"Алекс\"андр"` создаёт ту же строку, что и `'Алекс"андр'`. Нужно экранировать конкретный внутренний символ, чтобы сохранить внешнюю границу литерала.

**Ожидаемый вывод**

```text
Программирование на Python
Алекс"андр
PY-1'01
<class 'str'>
```

````python
# Задаём обычную строку с названием курса для сравнения.
course_name = "Программирование на Python"
# Ограничиваем имя одинарными кавычками. Двойная кавычка между буквами остаётся частью
# имени.
teacher_name = 'Алекс"андр'
# Ограничиваем код двойными кавычками. Одинарная кавычка внутри входит в текст кода.
course_code = "PY-1'01"

# Печатаем название курса на первой строке.
print(course_name)
# Печатаем имя вместе с его внутренней двойной кавычкой.
print(teacher_name)
# Печатаем код курса с внутренней одинарной кавычкой.
print(course_code)

# Печатаем класс строки course_name. Вид ограничителя никак не меняет строковый тип.
print(type(course_name))
````

**Пример 10. Цифры внутри строки**

Значение `'123'` состоит из трёх символов `'1'`, `'2'` и `'3'`. Кавычки создают строку, даже если каждый символ можно прочитать как цифру. Числовой литерал `123` создал бы значение типа `int`.

Программа печатает название, код и `str1`. Внешне строка `'123'` после печати выглядит как число, поскольку `print()` показывает её содержимое без ограничителей. Две проверки `type()` дают `<class 'str'>` и для названия курса, и для последовательности цифр.

Для строк `'123' + '4'` даёт `'1234'`. Числовая операция `int('123') + 4` даёт `127`. В этой ячейке имя преподавателя получает значение, но присутствия значения в переменной недостаточно для вывода, поскольку отдельного вызова с `teacher_name` здесь нет.

**Ожидаемый вывод**

```text
Программирование на Python
PY-101
123
<class 'str'>
<class 'str'>
```

````python
# Готовим обычную текстовую строку, тип которой затем сравним с str1.
course_name = "Программирование на Python"
# Создаём ещё одну строку с внутренней двойной кавычкой. Эта переменная остаётся в памяти,
# но в вызовах print дальше отсутствует.
teacher_name = 'Алекс"андр'
# Сохраняем строковый код из букв, дефиса и цифр.
course_code = "PY-101"
# Сохраняем последовательность цифр как строку. Тип задают кавычки.
str1 = '123'

# Печатаем подготовленное название курса.
print(course_name)
# Печатаем код курса, пропуская переменную teacher_name.
print(course_code)
# Печатаем строку из цифр. В выводе видим 123 без внешних кавычек.
print(str1)

# Проверяем тип обычного текста. Получаем класс str.
print(type(course_name))
# Проверяем тип текста из цифр. Получаем тот же класс str.
print(type(str1))
````

**Пример 11. Булевы признаки**

Тип `bool` хранит два значения `True` и `False`. У каждого имени в примере есть смысловой вопрос. Активен ли курс, открыт ли набор, предусмотрен ли сертификат. Префиксы `is_` и `has_` помогают читать такие имена в условиях.

Значения записаны без кавычек и начинаются с заглавных букв. Первые три вызова печати показывают два истинных и одно ложное значение в порядке переменных. Последняя строка с `type()` показывает `<class 'bool'>`.

Запись `"False"` создаёт непустую строку. В условии `if "False":` такая строка даст истинный результат, а булево значение `False` даст ложный. Это важно, если признаки приходят из текстового файла или формы.

**Ожидаемый вывод**

```text
True
False
True
<class 'bool'>
```

````python
# Отмечаем, что курс активен. True — одно из двух значений типа bool.
is_course_active = True
# Отмечаем, что набор закрыт. Для этого используем логическое значение False.
is_enrollment_open = False
# Отмечаем, что курс предусматривает сертификат.
has_certificate = True

# Печатаем признак активности курса. Он отображается как True.
print(is_course_active)
# Печатаем признак открытого набора. Он отображается как False.
print(is_enrollment_open)
# Печатаем признак наличия сертификата. Он отображается как True.
print(has_certificate)

# Определяем и выводим класс значения, чтобы увидеть тип bool.
print(type(is_course_active))
````

## 1.3 Математические операции. Понятие syntax sugar;

**Пример 12. Сложение и вычитание в одном выражении**

Три переменные представляют исходное число студентов, число пришедших и число ушедших. Начинаем с `18`, прибавляем `4` и вычитаем `2`. Промежуточно получается `22`, в результате остаётся `20`.

У бинарных `+` и `-` одинаковый приоритет, поэтому выражение группируется слева направо как `(students_count + new_students) - left_students`. Присваивание связывает `current_students` с вычисленным результатом, а исходные имена продолжают указывать на первоначальные значения.

Все операнды имеют тип `int`, поэтому результат тоже целочисленный. При проверке сценария с числом ушедших больше общего количества программа вычислит отрицательное число. Ограничение, которое проверяло бы допустимость такого ввода, является отдельным условием предметной области.

**Ожидаемый вывод**

```text
20
```

````python
# Запоминаем исходное количество студентов до изменений.
students_count = 18
# Запоминаем, сколько студентов пришли дополнительно.
new_students = 4
# Запоминаем число ушедших как положительное количество. Уменьшение задаст оператор
# вычитания в формуле.
left_students = 2

# Прибавляем 4 к 18 и затем вычитаем 2. Сохраняем результат 20 под новым именем.
current_students = students_count + new_students - left_students

# Печатаем итоговое количество студентов.
print(current_students)
````

**Пример 13. Умножение, обычное деление и тип результата**

Двадцать занятий по полтора часа дают `20 * 1.5 = 30.0` часа. Операнд `lesson_duration` имеет тип `float`, поэтому произведение целого числа и дробного тоже имеет тип `float`.

При расписании из пяти занятий в неделю выражение `20 / 5` даёт `4.0` недели. Оператор `/` при делении двух значений `int` даёт `float` даже при делении нацело. Контрольный вызов `type(weeks_count)` показывает этот класс явно.

При `21 / 5` результат был бы `4.2`. Если нужно считать только полные недели, подойдёт целочисленное деление `//`. Для числа календарных недель, необходимых под все занятия, используется своя формула с округлением вверх. Делитель в обоих видах деления должен быть отличным от нуля.

**Ожидаемый вывод**

```text
30.0
4.0
<class 'float'>
```

````python
# Задаём количество занятий целым числом 20.
lessons_count = 20
# Задаём длительность одного занятия в часах дробным числом 1.5.
lesson_duration = 1.5

# Умножаем количество занятий на их длительность. Получаем 30.0 часа.
total_hours = lessons_count * lesson_duration

# Показываем общий объём курса в часах.
print(total_hours)

# Указываем, что в неделю запланировано пять занятий.
weekly_lessons = 5
# Делим 20 на 5 оператором /. Результат записывается как float со значением 4.0.
weeks_count = lessons_count / weekly_lessons

# Печатаем рассчитанное число недель.
print(weeks_count)
# Показываем тип результата деления, чтобы подтвердить, что это float.
print(type(weeks_count))
````

**Пример 14. Полные группы и остаток**

Из `23` студентов нужно сформировать группы по `5` человек. Целочисленное деление `23 // 5` даёт `4` полные группы. Остаток `23 % 5` равен `3` и показывает, сколько студентов остаётся после распределения.

Связь между результатами можно проверить формулой `23 = 4 * 5 + 3`. Первым выводится число групп, вторым — остаток. Для положительного делителя остаток лежит в диапазоне от нуля до числа на единицу меньше делителя.

Если число студентов меньше размера группы, полных групп будет ноль, а остаток совпадёт с числом студентов. Размер группы в этой задаче должен быть положительным, а при делении на ноль оба оператора вызовут `ZeroDivisionError`.

**Ожидаемый вывод**

```text
4
3
```

````python
# Задаём общее количество студентов в потоке.
students_count = 23
# Задаём, сколько студентов входит в одну полную группу.
group_size = 5

# Находим число полных групп через //. Частное для положительных операндов округляется
# вниз, поэтому получаем 4.
full_groups = students_count // group_size
# Находим остаток через %. После четырёх групп по пять человек остаются три.
remaining_students = students_count % group_size

# Печатаем количество полных групп.
print(full_groups)
# Печатаем число студентов, оставшихся после формирования полных групп.
print(remaining_students)
````

**Пример 15. Возведение в степень**

В выражении `base ** exponent` слева стоит основание, справа — показатель степени. Если класс имеет квадратную форму со стороной `6`, его площадь равна `6 ** 2 = 36`. При измерении стороны в метрах результат выражается в квадратных метрах.

Во второй формуле восьмая степень двойки означает произведение восьми двоек. Результат `2 ** 8 = 256`, например столько различных комбинаций дают восемь двоичных разрядов. Здесь обе операции дают целые значения.

Символ `^` в Python выполняет поразрядное исключающее ИЛИ и даст другой результат. Для степени используйте две звёздочки. Возведение положительного основания в отрицательную степень, например `2 ** -1`, даёт обратную величину `0.5`.

**Ожидаемый вывод**

```text
36
256
```

````python
# Задаём длину стороны квадратного класса.
classroom_length = 6

# Возводим длину стороны в квадрат оператором **. Получаем 36 квадратных единиц.
classroom_area = classroom_length ** 2

# Печатаем вычисленную площадь.
print(classroom_area)

# Возводим основание 2 в восьмую степень. Результат равен 256.
storage_levels = 2 ** 8

# Печатаем вычисленную степень двойки.
print(storage_levels)
````

**Пример 16. Унарный минус и унарный плюс**

Унарный оператор работает с одним операндом. В присваивании унарный минус применяется к числу `1500` и даёт `-1500`. После этого первый `print()` показывает отрицательное изменение баланса.

Выражение `-balance_change` вычисляет число с противоположным знаком. Поскольку исходное значение отрицательно, получается положительное `1500`. Выражение `+balance_change` для целого числа сохраняет числовое значение, поэтому последний вывод снова равен `-1500`.

Обе операции выполняются внутри вызовов печати. Чтобы сохранить результат смены знака под тем же именем, потребовалось бы явное присваивание `balance_change = -balance_change`.

**Ожидаемый вывод**

```text
-1500
1500
-1500
```

````python
# Создаём отрицательное целое число. Минус перед 1500 задаёт изменение баланса в сторону
# уменьшения.
balance_change = -1500

# Печатаем исходное изменение баланса со знаком минус.
print(balance_change)
# Меняем знак значения только в аргументе print и видим положительное число 1500.
print(-balance_change)
# Применяем унарный плюс к исходному значению. Для этого целого числа значение остаётся
# -1500.
print(+balance_change)
````

**Пример 17. Двоичная погрешность и округление**

`0.1` и `0.2` записаны десятичными дробями, но `float` хранит их двоичные приближения. После сложения ближайшим представимым результатом становится число, выводимое как `0.30000000000000004`.

Первый вызов `print()` показывает обычное представление результата. Во втором вызове `round(result, 2)` округляет число до двух десятичных разрядов и возвращает `float`, который выводится как `0.3`. Второй аргумент `round()` задаёт точность округления значения.

Чтобы показать ровно две цифры после точки, можно отформатировать значение через `f"{result:.2f}"` и получить строку `'0.30'`. Округление внутри `print()` оставляет сохранённое значение `result` прежним.

**Ожидаемый вывод**

```text
0.30000000000000004
0.3
```

````python
# Задаём первый операнд как float. Десятичная дробь 0.1 представляется в двоичном формате
# приближённо.
first_value = 0.1
# Задаём второй операнд как float. Десятичная дробь 0.2 тоже приближённая.
second_value = 0.2

# Складываем два приближения и сохраняем значение, получающееся после округления до
# представимого float.
result = first_value + second_value

# Печатаем необработанный результат и видим дополнительные цифры после 0.3.
print(result)
# Округляем результат до двух десятичных разрядов перед печатью. Сам result сохраняет
# исходное значение.
print(round(result, 2))
````

**Пример 18. Обычное и составное присваивание**

Сначала имя `students_count` связано с числом `18`. Строка `students_count = students_count + 3` сначала вычисляет правую часть со старым значением, затем связывает имя с новым значением `21`.

Выражение `students_count += 2` для целого числа даёт тот же итог, что и `students_count = students_count + 2`. Оно использует уже обновлённое значение `21`, поэтому следующий результат равен `23`.

`+=` — краткая запись составного присваивания. Целые числа неизменяемы, и операция связывает имя с итоговым числом. У изменяемых объектов вроде списков поведение может отличаться, поскольку составная операция иногда меняет сам объект.

**Ожидаемый вывод**

```text
21
23
```

````python
# Связываем имя students_count с начальным числом 18.
students_count = 18

# Находим текущее значение 18, прибавляем 3 и связываем это же имя с результатом 21.
students_count = students_count + 3

# Показываем значение после первого обновления.
print(students_count)

# Используем составное присваивание. К текущему числу 21 добавляем 2.
students_count += 2

# Показываем окончательное значение 23 после двух обновлений.
print(students_count)
````

## 1.4 Форматирование данных: %, str.format(), f-strings;

**Пример 19. Подстановка одного значения через %**

Если слева от `%` стоит строка с полями формата, Python выполняет форматирование. Слева задаётся шаблон, справа передаётся значение для подстановки. Например, `"Студент: %s" % student_name` создаёт новую строку, которая затем поступает в `print()`.

`%s` использует строковое представление объекта. `%d` выводит целое число в десятичной системе. `%f` выводит число с фиксированной точкой и по умолчанию с шестью знаками после точки, поэтому `4.8` становится текстом `4.800000`.

Точность можно задать явным образом. Шаблон `"%.1f" % course_rating` даст `'4.8'`. Назначение оператора `%` зависит от типов операндов. Для двух целых чисел он вычисляет остаток от деления.

**Ожидаемый вывод**

```text
Студент: Александр
Количество баллов: 85
Рейтинг курса: 4.800000
```

````python
# Подготавливаем строку с именем студента для подстановки через %s.
student_name = "Александр"
# Подготавливаем целый балл для подстановки через %d.
exam_score = 85
# Подготавливаем дробный рейтинг для подстановки через %f.
course_rating = 4.8

# Оператор % подставляет строку в поле %s. Затем print выводит собранное сообщение.
print("Студент: %s" % student_name)
# Поле %d оформляет значение 85 как десятичное целое и вставляет его в текст.
print("Количество баллов: %d" % exam_score)
# Поле %f оформляет 4.8 с шестью знаками после точки, поэтому видим 4.800000.
print("Рейтинг курса: %f" % course_rating)
````

**Пример 20. Шаблон с несколькими полями через %**

Шаблон содержит три поля `%s`, `%d` и `%.2f`. Справа от оператора `%` стоит кортеж из трёх значений. Позиции сопоставляются последовательно. Имя поступает в первое поле, балл во второе, рейтинг в третье.

Для рейтинга `4.85678` спецификация `%.2f` задаёт ровно две цифры после точки. При округлении до сотых следующая цифра `6` увеличивает сотые, и в строку попадает `4.86`. Исходный рейтинг продолжает храниться как `4.85678`.

Скобки с запятыми объединяют три аргумента в кортеж и позволяют записать его на нескольких строках. Количество переданных значений должно соответствовать количеству полей. Например, пропущенный рейтинг приведёт к `TypeError`.

**Ожидаемый вывод**

```text
Студент: Александр, баллы: 85, рейтинг: 4.86
```

````python
# Сохраняем имя, которое будет первым аргументом формата.
student_name = "Александр"
# Сохраняем целое число баллов для второго поля.
exam_score = 85
# Задаём рейтинг с пятью знаками после точки, чтобы показать округление при выводе.
course_rating = 4.85678

# Объявляем шаблон итоговой строки и открываем кортеж значений для подстановки.
message = "Студент: %s, баллы: %d, рейтинг: %.2f" % (
    # Передаём имя, балл и рейтинг именно в порядке полей %s, %d и %.2f.
    student_name, exam_score, course_rating
# Закрываем кортеж. После форматирования готовая строка присваивается переменной message.
)

# Печатаем одну строку, в которой уже находятся все три значения.
print(message)
````

**Пример 21. Автоматический порядок полей str.format**

Скобки `{}` отмечают поля подстановки в строковом шаблоне. Метод `.format()` вызывается у этого шаблона. При пустых полях аргументы расходуются слева направо, начиная с первого.

Порядок присваиваний имеет значение только для подготовки значений к моменту вызова. Хотя `exam_score` объявлен раньше `course_name`, вызов передаёт сначала имя студента, затем название курса и последним балл. Число `85` получает текстовое представление автоматически.

Шаблон можно сохранить отдельно и применять с разными аргументами. Если для трёх полей передать только два позиционных аргумента, метод не сможет найти значение для третьего поля и вызовет `IndexError`.

**Ожидаемый вывод**

```text
Студент: Александр, курс: Python, баллы: 85
```

````python
# Сохраняем имя студента для первого поля шаблона.
student_name = "Александр"
# Сохраняем число баллов для последнего поля шаблона.
exam_score = 85
# Сохраняем название курса для среднего поля шаблона.
course_name = "Python"

# Создаём шаблон с тремя полями и вызываем его метод format. Результат запишется под
# именем message.
message = "Студент: {}, курс: {}, баллы: {}".format(
    # Передаём аргументы в порядке появления полей, чтобы имя, курс и баллы попали к
    # соответствующим подписям.
    student_name, course_name, exam_score
# Закрываем вызов format после списка аргументов.
)

# Передаём в print уже собранную строку.
print(message)
````

**Пример 22. Номера полей и повтор аргумента в str.format**

Числа внутри фигурных скобок — индексы позиционных аргументов. `{0}` выбирает первый переданный аргумент, `{1}` — второй. Счёт начинается с нуля.

В первой строке `{0}` предшествует `{1}`, поэтому сначала показан студент. Во второй шаблон меняет положение полей на `{1}` и `{0}`. В обоих вызовах аргументы передаются в одном порядке, а вывод меняется из-за порядка полей.

Третий шаблон содержит две ссылки `{0}`. Методу достаточно одного аргумента, поскольку оба поля читают один и тот же элемент. Явные индексы удобны при повторе значений и перестановке полей. В пределах одного шаблона сохраняйте один режим нумерации, поскольку смешивание поля `{}` с полем `{0}` вызывает `ValueError`.

**Ожидаемый вывод**

```text
Студент Александр изучает курс Python
Курс Python, студент Александр
Александр зарегистрирован. Имя студента: Александр
```

````python
# Подготавливаем имя. В первых двух вызовах оно будет аргументом с индексом 0.
student_name = "Александр"
# Подготавливаем название курса. Оно будет аргументом с индексом 1.
course_name = "Python"

# Шаблон первой строки выводит аргументы по возрастанию индекса, сначала 0, затем 1.
print("Студент {0} изучает курс {1}".format(
    # Передаём имя первым и курс вторым. Поле {0} использует имя, поле {1} использует
    # курс.
    student_name, course_name
# Закрываем вызов format и внешний print. Печатается первая строка.
))

# Во втором шаблоне ставим поле {1} раньше поля {0}, поэтому первым выводится курс.
print("Курс {1}, студент {0}".format(
    # Передаём аргументы в прежнем порядке. Порядок вывода определяет шаблон.
    student_name, course_name
# Закрываем format и print. Печатается строка с курсом перед именем.
))

# В третьем шаблоне поле {0} встречается дважды, поэтому одно значение будет вставлено в
# два места.
print("{0} зарегистрирован. Имя студента: {0}".format(
    # Передаём только имя. Два поля с индексом 0 используют этот аргумент.
    student_name
# Закрываем оба вызова и выводим строку с двумя вхождениями имени.
))
````

**Пример 23. Подстановка переменных через f-строку**

Буква `f` перед открывающей кавычкой включает интерполяцию. При вычислении литерала Python находит поля в фигурных скобках, вычисляет выражение в каждом поле и вставляет его строковое представление.

В примере выражениями служат три имени. Они дают `Александр`, `Python` и `85`. Текст между полями, включая запятые и пробелы, переносится в результат. Сформированное сообщение сохраняется в переменной и затем выводится.

Значения подставляются в момент вычисления f-строки. Если позднее изменить `exam_score`, старое сообщение в `message` сохранит прежний балл. При удалении префикса `f` тот же литерал будет содержать имена вместе с фигурными скобками.

**Ожидаемый вывод**

```text
Студент: Александр, курс: Python, баллы: 85
```

**Пример 24. Арифметические выражения внутри f-строк**

В фигурных скобках f-строки можно писать выражения. Базовая цена курса получается как `20 * 1500 = 30000`. Величина скидки задана долей `0.1`, поэтому для показа процентов программа умножает эту величину на `100` и добавляет буквальный символ `%` после поля.

Для цены со скидкой значение `1 - discount` даёт долю к оплате `0.9`. Умножение `30000 * 0.9` даёт `27000.0`. Разница между `30000` и `27000.0` в выводе объясняется типами. Первая формула использует два целых числа, вторая — также `float`.

В поле вывода процентов нет спецификации формата, поэтому `0.1 * 100` печатается как `10.0`. Для заданного числа знаков можно написать `{discount:.0%}`, где формат `%` сам преобразует долю в проценты. Для денежных расчётов выбор округления и представления суммы стоит задавать явно.

**Ожидаемый вывод**

```text
Количество занятий: 20
Стоимость обучения: 30000
Размер скидки: 10.0%
Итоговая стоимость: 27000.0
```

````python
# Задаём общее число занятий как целое число 20.
lessons_count = 20
# Задаём цену одного занятия целым числом 1500.
lesson_price = 1500
# Сохраняем скидку как долю от полной цены. 0.1 соответствует десяти процентам.
discount = 0.1

# Вставляем количество занятий непосредственно по имени переменной.
print(f"Количество занятий: {lessons_count}")
# Вычисляем 20 * 1500 внутри поля. Результат 30000 подставляется как текст.
print(f"Стоимость обучения: {lessons_count * lesson_price}")
# Переводим 0.1 в проценты умножением на 100. Символ % после поля остаётся обычным
# текстом.
print(f"Размер скидки: {discount * 100}%")
# Умножаем полную цену на остаток 1 - discount. Из-за дробного множителя выводится
# 27000.0.
print(f"Итоговая стоимость: {lessons_count * lesson_price * (1 - discount)}")
````

**Пример 25. Точность, проценты и разделение разрядов**

После двоеточия внутри поля f-строки задаётся формат представления. В спецификациях `.1f`, `.2f` и `.3f` буква `f` задаёт фиксированную точку, а число после точки — точное количество десятичных знаков. Поэтому один рейтинг отображается как `4.9`, `4.86` и `4.857`.

Спецификация `.1%` масштабирует долю до процентов, показывает один знак после точки и добавляет `%`. Именно так `0.156` превращается в `15.6%`. В спецификации `,.2f` запятая включает разделители групп по три цифры, а `.2f` оставляет две цифры после точки, поэтому цена выводится как `12,345.68`.

Форматирование меняет строковое отображение. Исходные числовые значения по-прежнему доступны с прежней точностью. В этом примере запятая группирует разряды, а десятичным разделителем остаётся точка. Выбор такой записи задаёт спецификация формата.

**Ожидаемый вывод**

```text
Рейтинг: 4.9
Рейтинг: 4.86
Рейтинг: 4.857
Скидка: 15.6%
Стоимость: 12,345.68
```

````python
# Сохраняем рейтинг с пятью цифрами после точки, чтобы сравнить три степени точности
# вывода.
course_rating = 4.85678
# Сохраняем скидку долей 0.156. Формат процентов позже выполнит масштабирование.
discount = 0.156
# Сохраняем цену с пятью цифрами в целой части и четырьмя в дробной.
course_price = 12345.6789

# Округляем для отображения до одного знака после точки. Получается 4.9.
print(f"Рейтинг: {course_rating:.1f}")
# Округляем для отображения до сотых. Получается 4.86.
print(f"Рейтинг: {course_rating:.2f}")
# Показываем три знака после точки. Получается 4.857.
print(f"Рейтинг: {course_rating:.3f}")

# Формат .1% превращает долю в 15.6% и автоматически добавляет знак процента.
print(f"Скидка: {discount:.1%}")
# Включаем запятую между группами тысяч и округляем до сотых. Получается 12,345.68.
print(f"Стоимость: {course_price:,.2f}")
````

**Пример 26. Ширина поля и выравнивание строки**

Формат `15` выделяет строке поле шириной не меньше 15 символов. У слова `Python` шесть букв, поэтому для такого поля нужны девять пробелов. Вертикальные черты в шаблоне показывают реальные границы поля.

Для текстового поля выравнивание по умолчанию совпадает с `<`. Первые две строки содержат слово слева и девять пробелов справа. `>` даёт девять пробелов слева. `^` распределяет пробелы вокруг слова. В данном случае четыре пробела появляются перед ним, пять после него.

Ширина задаёт нижнюю границу. Строка длиннее 15 символов будет выведена целиком. Для ограничения длины текста потребуется отдельная операция, например срез.

**Ожидаемый вывод**

```text
|Python         |
|Python         |
|         Python|
|    Python     |
```

````python
# Задаём строку из шести символов, которую разместим в полях шириной 15.
course_name = "Python"

# Формируем поле с выравниванием строки по умолчанию. Слово слева, а справа девять
# пробелов.
print(f"|{course_name:15}|")
# Явно указываем левое выравнивание знаком <. Оно совпадает с предыдущим выводом.
print(f"|{course_name:<15}|")
# Выравниваем слово вправо знаком >. Девять пробелов появляются перед словом.
print(f"|{course_name:>15}|")
# Центрируем слово знаком ^. Здесь четыре пробела слева и пять справа.
print(f"|{course_name:^15}|")
````

**Пример 27. Символ заполнения и номера с ведущими нулями**

Спецификация `-^20` состоит из символа заполнения `-`, центрального выравнивания `^` и ширины `20`. В слове `Python` шесть символов, поэтому четырнадцать свободных мест заполняются дефисами, по семь с каждой стороны.

Формат `05` для целого числа включает дополнение нулями до ширины 5. У числа `42` две цифры, поэтому перед ним появляются три нуля и выводится `00042`.

Ведущие нули относятся к отображению номера. Числовое значение в `student_number` остаётся равным `42`. Если требуется хранить идентификатор именно в виде `'00042'`, можно сохранить отдельно результат форматирования.

**Ожидаемый вывод**

```text
-------Python-------
00042
```

````python
# Задаём строку, которую разместим по центру поля.
course_name = "Python"
# Задаём номер студента как целое число 42.
student_number = 42

# Центрируем Python в поле шириной 20 и заполняем свободные позиции дефисами.
print(f"{course_name:-^20}")
# Показываем номер в поле шириной 5 с ведущими нулями. Получаем 00042.
print(f"{student_number:05}")
````

**Пример 28. Отладочный синтаксис и буквальные скобки**

Знак `=` после выражения внутри поля f-строки включает отладочную форму. Шаблон `{students_count=}` выводит исходную запись выражения, знак равенства и представление значения. Удобно видеть сразу имя и его текущее значение.

В третьем поле вычисляется произведение `18 * 1500`. В результате остаётся исходная запись `students_count * lesson_price` с пробелами вокруг оператора и добавляется результат `27000`.

Фигурные скобки в текстовой части f-строки удваивают, чтобы получить буквальный символ. `{{` печатает `{`, а `}}` печатает `}`. Пара `{{}}` превращается в `{}`. Одиночная пустая пара `{}` в f-строке считается пустым полем и вызывает синтаксическую ошибку.

## 1.5 Приоритеты операций;

**Пример 29. Умножение перед сложением**

В выражении `10 + 5 * 2` умножение имеет более высокий приоритет, чем сложение. Python группирует его как `10 + (5 * 2)`. Сначала вычисляется `5 * 2 = 10`, затем `10 + 10 = 20`.

Знак `=` связывает имя `result` с целым числом `20` после вычисления правой части. Следующий вызов печатает уже сохранённый итог. Расположение `+` раньше `*` в тексте не меняет правил группировки.

Чтобы посчитать сумму до умножения, можно явно поставить скобки вокруг `10 + 5`. Тогда выражение `(10 + 5) * 2` даст `30`.

**Ожидаемый вывод**

```text
20
```

````python
# Сначала вычисляем произведение 5 * 2, затем прибавляем его к 10 и сохраняем целое число
# 20.
result = 10 + 5 * 2

# Печатаем сохранённый результат вычисления.
print(result)
````

**Пример 30. Скобки меняют порядок арифметики**

Два выражения используют одинаковые числа и операции. В `10 + 5 * 2` произведение `5 * 2` вычисляется внутри сложения, поэтому получается `20`. В `(10 + 5) * 2` скобки делают результат сложения первым операндом умножения.

Во втором случае сначала получаем `10 + 5 = 15` и затем удваиваем эту величину. Вывод `20` и `30` позволяет увидеть разницу, вызванную группировкой.

Скобки полезны для избежания двусмысленности. Запись `10 + (5 * 2)` дала бы те же `20`, поскольку она явно повторяет стандартный приоритет.

**Ожидаемый вывод**

```text
20
30
```

````python
# Используем обычный приоритет. Умножение 5 * 2 даёт 10, после сложения получаем 20.
result_1 = 10 + 5 * 2
# Группируем сложение скобками. Умножаем результат 15 на 2 и получаем 30.
result_2 = (10 + 5) * 2

# Печатаем результат выражения без скобок.
print(result_1)
# Печатаем результат выражения со скобками вокруг суммы.
print(result_2)
````

**Пример 31. Левая ассоциативность деления и умножения**

Операторы `/` и `*` имеют одинаковый приоритет. Для этих операторов выражение без скобок группируется слева направо. `20 / 5 * 2` читается как `(20 / 5) * 2`. После деления получаем `4.0`, после умножения — `8.0`.

В выражении `20 / (5 * 2)` скобки меняют структуру. Делителем становится произведение `10`, поэтому результат равен `2.0`. В обоих случаях обычное деление `/` возвращает `float`, даже когда дробная часть равна нулю.

Если в формуле знаменатель состоит из произведения, заключайте всё произведение в скобки. Иначе второй множитель будет умножать уже полученное частное.

**Ожидаемый вывод**

```text
8.0
2.0
```

````python
# Группируем операции слева направо, потому что / и * равноприоритетны. Делим 20 на 5 и
# удваиваем 4.0.
result_1 = 20 / 5 * 2
# Сначала находим делитель 5 * 2 в скобках. Деление 20 на 10 даёт 2.0.
result_2 = 20 / (5 * 2)

# Печатаем результат деления с последующим умножением.
print(result_1)
# Печатаем результат деления на произведение.
print(result_2)
````

**Пример 32. Правая ассоциативность степеней**

Два соседних оператора `**` группируются справа налево. Поэтому в `2 ** 3 ** 2` показателем внешней степени становится `3 ** 2`. Этот показатель равен `9`, а `2 ** 9` даёт `512`.

Второе присваивание явно ставит скобки вокруг показателя и тем самым повторяет группировку первого. Оба вывода совпадают и показывают `512`.

Можно проверить другую группировку вычислением `(2 ** 3) ** 2`. В таком случае внутренний результат `8` возводится в квадрат и получается `64`. Эта разница объясняет, почему при чтении цепочек степеней важно учитывать ассоциативность.

**Ожидаемый вывод**

```text
512
512
```

````python
# Используем правую ассоциативность **. Сначала вычисляется 3 ** 2, затем 2 ** 9.
result_1 = 2 ** 3 ** 2
# Записываем ту же группировку с явными скобками и получаем те же 512.
result_2 = 2 ** (3 ** 2)

# Выводим результат цепочки степеней без скобок.
print(result_1)
# Выводим результат степени с явно сгруппированным показателем.
print(result_2)
````

**Пример 33. Унарный минус и отрицательное основание степени**

В `-3 ** 2` оператор степени связывается с числом `3` раньше унарного минуса, стоящего слева. Вычисление эквивалентно `-(3 ** 2)`. Сначала получается `9`, затем применяется знак минус и итог равен `-9`.

В `(-3) ** 2` скобки делают отрицательное число основанием. Квадрат отрицательного числа равен произведению двух чисел с отрицательным знаком, поэтому получается `9`.

Отдельный случай — отрицательный показатель справа. Выражение `2 ** -1` читается как `2 ** (-1)` и даёт `0.5`. Для отрицательного основания, особенно рядом со степенью, скобки делают намерение явным.

**Ожидаемый вывод**

```text
-9
9
```

````python
# Вычисляем квадрат положительного числа 3. Унарный минус применяется к результату 9.
result_1 = -3 ** 2
# Делаем отрицательное число -3 основанием степени с помощью скобок. Его квадрат равен 9.
result_2 = (-3) ** 2

# Печатаем -9 как результат применения минуса после возведения в степень.
print(result_1)
# Печатаем 9 как квадрат отрицательного основания.
print(result_2)
````

## 1.6 Ссылочная модель в Python, функция id(), оператор is;

**Пример 34. Два имени для одного целого числа**

Оператор присваивания связывает имя слева со значением выражения справа. Сначала имя `students_count` связывается с целочисленным объектом со значением `20`. При выполнении второго присваивания Python находит объект по имени `students_count` и связывает с этим объектом ещё одно имя, `current_students`.

Первый вызов `print()` получает объект через имя `students_count`, второй вызов получает тот же объект через имя `current_students`. Оба раза печатается строковое представление числа `20`. Второе присваивание обошлось без копирования объекта и без повторного вычисления числа.

По двум одинаковым строкам вывода можно судить только о напечатанном представлении. Общая идентичность в этом примере следует из самого присваивания `current_students = students_count`. Если позже присвоить новый объект одному имени, второе имя сохранит свою прежнюю связь.

**Ожидаемый вывод**

```text
20
20
```

````python
# Связываем имя students_count с целочисленным объектом со значением 20.
students_count = 20
# Читаем объект через имя students_count и связываем с тем же объектом имя
# current_students.
current_students = students_count

# Передаём print() число, доступное по имени students_count, и получаем 20.
print(students_count)
# Обращаемся к тому же числу по второму имени и снова получаем 20.
print(current_students)
````

**Пример 35. Проверка идентичности через id()**

Функция `id()` возвращает целочисленный идентификатор объекта. Идентификатор сохраняется в течение жизни объекта. Два разных объекта, которые существуют одновременно, имеют разные идентификаторы.

Присваивание `current_students = students_count` даёт объекту со значением `20` второе имя. Сначала `id()` получает объект по первому имени, затем по второму. Поэтому оба вызова возвращают один идентификатор, а `print()` выводит его дважды.

Число, возвращённое `id()`, зависит от запуска. CPython использует адрес объекта в памяти, другие реализации могут выбирать идентификатор иначе. После завершения жизни объекта его идентификатор может получить другой объект, поэтому равенство значений `id()`, зафиксированных в разное время, требует учитывать время жизни обоих объектов.

**Пример вывода с условными идентификаторами**

```text
ID_A
ID_A
```

`ID_A` обозначает одно и то же число во всех строках. Конкретный идентификатор зависит от запуска.

````python
# Связываем имя students_count с объектом целого числа 20.
students_count = 20
# Даём этому же объекту второе имя, current_students.
current_students = students_count

# Запрашиваем у id() идентификатор объекта, связанного с students_count, и печатаем его.
print(id(students_count))
# Запрашиваем идентификатор через второе имя. Функция получает тот же объект и возвращает
# то же число.
print(id(current_students))
````

**Пример 36. Перепривязка имени и сравнение идентификаторов**

После двух первых присваиваний оба имени указывают на объект `20`. Первые два вызова `print(id(...))` закрепляют начальное наблюдение. В этой паре идентификаторы совпадают, поскольку оба обращения ведут к одному объекту.

Строка `students_count = 25` связывает только имя слева с объектом со значением `25`. Имя `current_students` по-прежнему связано с объектом `20`. Поскольку оба объекта существуют одновременно и имеют разные значения, следующая пара идентификаторов различается. Идентификатор объекта по имени `current_students` совпадает с первоначальным.

Последние вызовы `print()` показывают значения, доступные по именам после перепривязки. Сначала печатается `25`, затем `20`. Целые числа в Python неизменяемы. Когда справа появляется другое число, присваивание изменяет связь имени с объектом.

**Пример вывода с условными идентификаторами**

```text
ID_A
ID_A
ID_B
ID_A
25
20
```

`ID_A` и `ID_B` обозначают два разных идентификатора. Конкретные числа зависят от запуска.

````python
# Связываем имя students_count с целым числом 20.
students_count = 20
# Читаем исходный объект и связываем с ним имя current_students.
current_students = students_count

# Печатаем первоначальный идентификатор объекта по имени students_count.
print(id(students_count))
# Печатаем тот же первоначальный идентификатор через имя current_students.
print(id(current_students))

# Перепривязываем только students_count к объекту со значением 25.
students_count = 25

# Печатаем идентификатор объекта 25, доступного по имени students_count.
print(id(students_count))
# Печатаем первоначальный идентификатор объекта 20, который всё ещё доступен по имени
# current_students.
print(id(current_students))

# Читаем значение по перепривязанному имени students_count и выводим 25.
print(students_count)
# Читаем значение по имени current_students и выводим 20.
print(current_students)
````

**Пример 37. Проверка тождественности оператором is**

Выражение `a is b` даёт `True`, если вычисление обоих операндов приводит к одному объекту. Для такой проверки можно обратиться к именам прямо, без вызова `id()`. В примере первый результат равен `True`, потому что `current_students` получил объект, ранее связанный с `students_count`.

После присваивания `students_count = 25` первое имя указывает на объект `25`, второе имя сохраняет связь с объектом `20`. Один объект целого числа имеет одно значение. Здесь объекты заведомо различны, поэтому второе выражение с `is` даёт `False`.

Для проверки равенства учебных баллов, количества студентов и других числовых значений используют `==`. Применение `is` оправдано, когда важно наличие конкретного объекта, например при проверке `value is None`.

**Ожидаемый вывод**

```text
True
False
```

````python
# Связываем имя students_count с числом 20.
students_count = 20
# Связываем имя current_students с уже найденным объектом 20.
current_students = students_count

# Сравниваем идентичность объектов по двум именам и печатаем True.
print(students_count is current_students)

# Связываем students_count с объектом со значением 25. По имени current_students
# по-прежнему доступно 20.
students_count = 25

# Повторяем проверку идентичности после перепривязки. Для объектов 25 и 20 результат равен
# False.
print(students_count is current_students)
````

**Пример 38. Равенство значений и тождественность объектов**

Каждый вызов `int("1000")` читает цифры строки и получает целое число `1000`. Два вызова сделаны независимо. Для целых чисел оператор `==` проверяет равенство числовых значений, поэтому первый результат равен `True`.

Оператор `is` проверяет идентичность объектов. В CPython эти два преобразования строки со значением `1000` обычно дают два объекта, и второй результат равен `False`. В этом примере одинаковый числовой результат получен через независимые вызовы, что позволяет увидеть различие двух проверок.

Язык Python допускает повторное использование объектов неизменяемых типов. На результат проверки идентичности могут влиять реализация и способ получения числа, например использование ранее созданного объекта или заранее сохранённой константы. Для равенства любых двух чисел следует полагаться на `==`.

**Пример вывода в CPython**

```text
True
False
```

````python
# Преобразуем строку с цифрами в целое число 1000 и связываем результат с number_1.
number_1 = int("1000")
# Независимо выполняем то же преобразование и сохраняем результат по имени number_2.
number_2 = int("1000")

# Сравниваем числовые значения. Они равны, поэтому печатается True.
print(number_1 == number_2)
# Проверяем, совпадают ли сами объекты. В CPython при таком получении числа обычно
# печатается False.
print(number_1 is number_2)
````

**Пример 44. Округление вниз, вверх и к нулю**

Функция `floor(x)` возвращает наибольшее целое, которое меньше либо равно `x`, а `ceil(x)` возвращает наименьшее целое, которое больше либо равно `x`. Функция `trunc(x)` отбрасывает дробную часть. Для аргументов типа `float` все три функции возвращают целое число типа `int`.

У положительного `4.7` соседние целые на числовой прямой равны `4` и `5`. Округление вниз даёт `4`, вверх даёт `5`, отбрасывание дробной части даёт `4`. Именно такой порядок соответствует первым трём вызовам `print()`.

У отрицательного `-4.7` меньший сосед равен `-5`, больший сосед равен `-4`. После перепривязки `number` функции выводят `-5`, `-4` и `-4`. У отрицательных дробных чисел `trunc()` движется к нулю, поэтому её результат совпадает с `ceil()`.

**Ожидаемый вывод**

```text
4
5
4
-5
-4
-4
```

````python
# Импортируем модуль, чтобы использовать три способа преобразования дробного числа в
# целое.
import math

# Начинаем сравнение с положительного числа 4.7.
number = 4.7

# Берём ближайшее целое снизу и для 4.7 получаем 4.
print(math.floor(number))
# Берём ближайшее целое сверху и для 4.7 получаем 5.
print(math.ceil(number))
# Отбрасываем дробную часть положительного числа и получаем 4.
print(math.trunc(number))

# Заменяем значение number на отрицательное число той же абсолютной величины.
number = -4.7

# Округляем −4.7 вниз на числовой прямой и получаем −5.
print(math.floor(number))
# Округляем −4.7 вверх на числовой прямой и получаем −4.
print(math.ceil(number))
# Отбрасываем дробную часть в направлении нуля. Для −4.7 это даёт −4.
print(math.trunc(number))
````

**Пример 45. Радианы и тригонометрические функции**

Угол `45` первоначально задан в градусах, а `math.sin()`, `math.cos()` и `math.tan()` принимают аргументы в радианах. Преобразование умножает число градусов на π/180, поэтому `math.radians(45)` возвращает вещественное приближение к π/4.

Для угла π/4 синус и косинус математически равны √2/2, то есть примерно `0.7071`. Тангенс равен отношению синуса к косинусу и математически равен `1`. Последние цифры напечатанных значений могут различаться из-за приближённого представления аргумента и конечной точности вычислений.

В конце `math.degrees(radians)` переводит число радиан обратно в градусы, что для данного угла даёт `45.0`. Если вызвать `math.sin(45)` прямо, функция обработает угол в сорок пять радиан и вычислит другое значение. Python принимает оба аргумента как числа, поэтому согласование единиц измерения лежит на авторе программы.

**Ожидаемый вывод**

```text
0.7071067811865476
0.7071067811865476
0.9999999999999999
45.0
```

````python
# Импортируем функции для преобразования углов и вычисления тригонометрических значений.
import math

# Сохраняем исходный угол в градусах.
angle = 45
# Переводим 45 градусов в радианы и сохраняем приближение к π/4.
radians = math.radians(angle)

# Печатаем синус угла в радианах, приблизительно 0.7071.
print(math.sin(radians))
# Печатаем косинус того же угла, приблизительно 0.7071.
print(math.cos(radians))
# Печатаем тангенс угла. Вещественный результат близок к 1.0.
print(math.tan(radians))

# Переводим сохранённые радианы обратно в градусы и печатаем 45.0.
print(math.degrees(radians))
````

**Пример 46. Факториал, общие делители, кратные и близость чисел**

Факториал `5!` равен произведению `1 × 2 × 3 × 4 × 5`, поэтому `math.factorial(5)` возвращает `120`. Для факториала требуется неотрицательное целое число. Отдельный граничный случай, `0!`, равен единице.

Функция `math.gcd(24, 36)` возвращает наибольший общий делитель `12`. Оба аргумента делятся на него без остатка, и среди их общих делителей он наибольший. Функция `math.lcm(4, 6)` возвращает наименьшее положительное общее кратное `12`, поскольку это первое положительное число, делящееся на `4` и `6`. В `math.fabs(-15)` вычисляется абсолютная величина, поэтому результат имеет вид `15.0`.

Числа `0.1`, `0.2` и `0.3` хранятся в типе `float` с конечной двоичной точностью. Сумма первых двух получается очень близкой к третьему числу. `math.isclose()` проверяет, укладывается ли разница в заданный допуск, и в этом примере с допуском по умолчанию даёт `True`. Для сравнения очень маленьких чисел с нулём может понадобиться положительный абсолютный допуск `abs_tol`.

**Ожидаемый вывод**

```text
120
12
12
15.0
True
```

````python
# Импортируем функции для дискретных вычислений, абсолютной величины и сравнения дробных
# чисел.
import math

# Перемножаем целые числа от 1 до 5 и печатаем факториал 120.
print(math.factorial(5))
# Находим наибольшее целое число, на которое делятся и 24, и 36, и печатаем 12.
print(math.gcd(24, 36))
# Находим наименьшее положительное число, которое делится и на 4, и на 6, и печатаем 12.
print(math.lcm(4, 6))
# Находим абсолютную величину числа −15 и печатаем результат 15.0 типа float.
print(math.fabs(-15))
# Сравниваем сумму 0.1 и 0.2 с числом 0.3 с учётом допустимой погрешности. Результат равен
# True.
print(math.isclose(0.1 + 0.2, 0.3))
````

**Пример 47. Шесть операторов сравнения чисел**

Переменные хранят результат экзамена `75` и проходной балл `60`. В каждом выражении слева стоит фактический балл, справа стоит порог. Для этих чисел операторы сравнения возвращают значения типа `bool`. `print()` печатает их как `True` и `False`.

Поскольку `75` больше `60`, выражения `>` и `>=` истинны, а выражения `<` и `<=` ложны. Два числа различаются, поэтому `==` даёт `False`, а `!=` даёт `True`. Каждая строка вычисляет своё сравнение отдельно.

Разница между строгим и нестрогим сравнениями проявляется на границе. Если оба балла равны `60`, выражение `>=` истинно, а `>` ложно. Два знака равенства образуют оператор сравнения `==`. Один знак `=` в строках выше использован для присваивания.

**Ожидаемый вывод**

```text
True
False
True
False
False
True
```

````python
# Сохраняем фактический результат экзамена, равный 75 баллам.
exam_score = 75
# Записываем проходной балл, с которым сравниваются 75 баллов.
passing_score = 60

# Проверяем, строго ли 75 больше 60. Результат равен True.
print(exam_score > passing_score)
# Проверяем, строго ли 75 меньше 60. Результат равен False.
print(exam_score < passing_score)
# Проверяем, достигнут ли порог 60 с учётом возможности равенства. Для 75 результат равен
# True.
print(exam_score >= passing_score)
# Проверяем, меньше ли 75 порога 60 или равно ли ему. Получаем False.
print(exam_score <= passing_score)
# Проверяем точное равенство двух целых чисел. Значения различаются, поэтому печатается
# False.
print(exam_score == passing_score)
# Проверяем неравенство двух целых чисел. У 75 и 60 разные значения, поэтому печатается
# True.
print(exam_score != passing_score)
````

**Пример 48. Выполнение блока if при истинном условии**

У оператора `if` за ключевым словом следует условие, затем в конце строки ставится двоеточие. Python вычисляет `exam_score >= passing_score`. При значениях `75` и `60` сравнение даёт `True`, поэтому управление переходит в блок с отступом.

Оба вызова `print()` имеют одинаковый отступ относительно заголовка `if` и принадлежат его телу. Сначала печатается сообщение о сдаче экзамена, затем сообщение о допуске к следующему модулю. Знаки `>=` включают проходной балл в диапазон успешных результатов.

Последний вызов `print()` возвращается на уровень отступа заголовка `if` и выполняется после конструкции. При балле `59` условие стало бы ложным, Python пропустил бы обе вложенные строки и всё равно сообщил бы о завершении проверки. Случайный отступ у последней строки сделал бы и это сообщение условным.

**Ожидаемый вывод**

```text
Экзамен успешно сдан
Студент допущен к следующему модулю
Проверка результатов завершена
```

````python
# Записываем фактический результат в 75 баллов.
exam_score = 75
# Сохраняем минимальный балл для сдачи экзамена.
passing_score = 60

# Проверяем, достиг ли результат проходного балла. Сравнение 75 и 60 даёт True.
if exam_score >= passing_score:
    # В первой строке тела if сообщаем об успешной сдаче.
    print("Экзамен успешно сдан")
    # На том же уровне отступа сообщаем о допуске к следующему модулю.
    print("Студент допущен к следующему модулю")

# Возвращаемся к отступу заголовка if и сообщаем о завершении проверки при любом
# результате сравнения.
print("Проверка результатов завершена")
````

**Пример 49. Выбор одной из двух ветвей if и else**

Результат экзамена равен `45`, а проходной балл равен `60`. Сравнение `45 >= 60` даёт `False`. Python пропускает блок `if` и выполняет блок `else`, который задаёт действия для всех случаев, когда условие ложно.

Заголовок `else` состоит из ключевого слова и двоеточия. Заголовки `if` и `else` стоят на одном уровне отступа, а строки внутри каждой ветви имеют дополнительный отступ. В данном запуске печатаются оба сообщения из `else`, сначала о результате экзамена, затем о пересдаче.

После выбранной ветви Python продолжает выполнение первой строки с отступом заголовков. Сообщение о завершении проверки появляется и для `45`, и для `60`. При ровно `60` срабатывает ветвь успешной сдачи, потому что `>=` включает равенство.

**Ожидаемый вывод**

```text
Экзамен не сдан
Необходимо пройти пересдачу
Проверка результатов завершена
```

````python
# Записываем результат в 45 баллов, чтобы пройти по ветви пересдачи.
exam_score = 45
# Задаём проходное значение 60 баллов.
passing_score = 60

# Сравниваем 45 с проходным баллом. Условие даёт False.
if exam_score >= passing_score:
    # Эта строка сообщает об успешной сдаче и выполняется только при истинном условии if.
    print("Экзамен успешно сдан")
# Задаём ветвь для ложного результата сравнения. В этом запуске Python входит в неё.
else:
    # Печатаем первое сообщение ветви else о результате экзамена.
    print("Экзамен не сдан")
    # Следующей строкой той же ветви выводим необходимость пересдачи.
    print("Необходимо пройти пересдачу")

# После любой выбранной ветви печатаем сообщение о завершении проверки.
print("Проверка результатов завершена")
````

**Пример 50. Выбор оценки по первому подходящему порогу**

Цепочка `if`, `elif` и `else` перебирает взаимоисключающие варианты действий сверху вниз. Сначала Python проверяет, достигает ли балл порога `90`. Для `85` ответ отрицательный, поэтому проверяется следующий порог, `75`.

Условие `85 >= 75` истинно. Python печатает «Оценка: хорошо» и заканчивает всю цепочку. Сравнение с `60` могло бы дать `True`, если бы вычислялось отдельно. В этой цепочке до него управление уже не доходит.

Порядок порогов определяет итоговую оценку. При целых баллах ветви соответствуют диапазонам от `90` и выше, от `75` до `89`, от `60` до `74` и ниже `60`. Если переместить проверку `>= 60` наверх, баллы `75` и `90` тоже попадут в неё и получат оценку «удовлетворительно».

**Ожидаемый вывод**

```text
Оценка: хорошо
```

````python
# Задаём балл 85, который находится между порогами 75 и 90.
exam_score = 85

# Начинаем с самого высокого порога. Для 85 условие ложно.
if exam_score >= 90:
    # Печатаем высшую оценку при балле от 90. При текущих данных эта строка пропускается.
    print("Оценка: отлично")
# После ложного результата первой проверки сравниваем балл с 75. Для 85 это условие
# истинно.
elif exam_score >= 75:
    # Печатаем выбранную оценку «хорошо». Оставшаяся часть цепочки пропускается.
    print("Оценка: хорошо")
# Этот порог проверяется, когда балл ниже 75. Порядок условий задаёт верхнюю границу
# диапазона.
elif exam_score >= 60:
    # Сообщаем об удовлетворительной оценке при балле от 60 до 74 включительно.
    print("Оценка: удовлетворительно")
# Обрабатываем случай, когда ни один из трёх порогов не достигнут.
else:
    # Печатаем неудовлетворительную оценку для балла ниже 60.
    print("Оценка: неудовлетворительно")
````

**Пример 51. Независимые проверки для накопительного результата**

Здесь каждый `if` начинает отдельную конструкцию. После проверки `exam_score >= 60` выполнение продолжается у следующего `if` даже тогда, когда тело первой проверки уже выполнено. Все три порога проверяются по порядку.

Баллу `95` соответствуют все три условия. Сначала выводится сообщение о сдаче экзамена, затем сообщение о хорошем результате, затем рекомендация к поощрению. При значении `80` сработали бы только первые два условия, а при `59` ни одного сообщения не было бы.

Такая форма подходит, когда более высокий балл даёт несколько результатов одновременно. Если нужна ровно одна итоговая категория, цепочка `if` и `elif` останавливается после первой подходящей ветви. При переводе этих порогов в такую цепочку для итоговой категории следовало бы начинать с `90`.

**Ожидаемый вывод**

```text
Экзамен успешно сдан
Студент показал хороший результат
Студент рекомендован к поощрению
```

````python
# Записываем высокий результат в 95 баллов, который проходит все три порога.
exam_score = 95

# Выполняем самостоятельную проверку минимального проходного балла 60.
if exam_score >= 60:
    # При достижении 60 баллов сообщаем об успешной сдаче.
    print("Экзамен успешно сдан")

# Вторым независимым условием проверяем достижение 75 баллов.
if exam_score >= 75:
    # При выполнении второго условия добавляем сообщение о хорошем результате.
    print("Студент показал хороший результат")

# Третьим независимым условием проверяем порог поощрения в 90 баллов.
if exam_score >= 90:
    # Так как 95 достигает порога 90, выводим ещё и рекомендацию к поощрению.
    print("Студент рекомендован к поощрению")
````

**Пример 52. Вложенная проверка оплаты и экзамена**

Внешний `if` определяет, оплачено ли обучение. Имя `is_course_paid` уже связано с логическим значением `True`, поэтому Python сразу использует его как условие и входит в тело внешней ветви. Внутри запускается отдельная проверка проходного балла.

Условие `85 >= 60` истинно, поэтому выводится «Сертификат доступен». Если обучение оплачено, а балл меньше `60`, срабатывает вложенный `else` и появляется сообщение о необходимости сдать экзамен. При отсутствии оплаты срабатывает внешний `else`, а сравнение балла с порогом пропускается.

Принадлежность каждого `else` определяет отступ его заголовка. Вложенный `else` выровнен с `if exam_score >= 60`, а внешний выровнен с `if is_course_paid`. Отступы кодируют порядок проверок. Даже при высоком балле неоплаченный студент получит сообщение об оплате.

**Ожидаемый вывод**

```text
Сертификат доступен
```

````python
# Сохраняем результат экзамена, равный 85 баллам.
exam_score = 85
# Задаём явное логическое значение True, которое означает, что обучение оплачено.
is_course_paid = True

# Проверяем оплату перед проверкой экзамена. При True переходим к вложенному условию.
if is_course_paid:
    # Находимся внутри оплаченной ветви и проверяем, достаточно ли 85 баллов.
    if exam_score >= 60:
        # При выполнении обеих проверок сообщаем, что сертификат доступен.
        print("Сертификат доступен")
    # Этот else выровнен с вложенным if и обрабатывает балл ниже проходного.
    else:
        # Сообщаем о необходимости сдать экзамен студенту, который уже оплатил обучение.
        print("Необходимо сдать экзамен")
# Этот else выровнен с внешним if и обрабатывает отсутствие оплаты.
else:
    # Выводим требование оплатить обучение независимо от результата экзамена.
    print("Необходимо оплатить обучение")
````

**Пример 53. Присваивание статуса в обеих ветвях**

Условие `exam_score >= 60` выбирает строковое значение для имени `exam_status`. При балле `85` условие равно `True`, поэтому имя связывается со строкой `"Сдан"`.

При балле ниже `60` управление переходит в `else`, где то же имя связывается со строкой `"Не сдан"`. Благодаря присваиванию в каждой ветви статус определён к моменту печати при любом исходе проверки.

Последний `print()` выполняется за пределами ветвей и читает выбранное значение один раз. Если убрать присваивание из одной ветви и запустить этот случай в свежем ядре, чтение имени на верхнем уровне даст `NameError`. В уже использованном ядре имя могло сохраниться после предыдущего запуска и дать устаревший результат.

**Ожидаемый вывод**

```text
Сдан
```

````python
# Устанавливаем фактический результат экзамена в 85 баллов.
exam_score = 85

# Проверяем, входит ли этот результат в диапазон успешной сдачи от 60 баллов.
if exam_score >= 60:
    # При истинном условии связываем exam_status со строкой «Сдан».
    exam_status = "Сдан"
# Обрабатываем все баллы ниже 60 во второй ветви.
else:
    # При ложном условии связываем то же имя со строкой «Не сдан».
    exam_status = "Не сдан"

# Один раз печатаем статус после того, как выбранная ветвь присвоила ему значение.
print(exam_status)
````

**Пример 54. Выбор значения условным выражением**

Условное выражение записывается в виде `значение_если_истина if условие else значение_если_ложь`. Хотя первая строка-кандидат записана левее условия, Python начинает с вычисления условия `exam_score >= 60`.

При значении `85` условие истинно. Python вычисляет левое выражение и получает строку `"Сдан"`. Правая строка выбирается только для балла меньше `60`. Выбранное значение становится результатом всего выражения и присваивается имени `exam_status`.

Обе ветви представляют собой выражения, из которых выбирается ровно одно. Поэтому такой синтаксис хорошо читается для двух коротких значений. Если в ветви требуется выполнить набор самостоятельных инструкций, его записывают блоком обычного `if` и `else`.

**Ожидаемый вывод**

```text
Сдан
```

````python
# Сохраняем исходный балл 85 для расчёта статуса.
exam_score = 85

# Сначала проверяем порог 60, затем выбираем одну строку. Для 85 в exam_status
# записывается «Сдан».
exam_status = "Сдан" if exam_score >= 60 else "Не сдан"

# Печатаем выбранный условным выражением статус.
print(exam_status)
````

**Пример 55. Расчёт цены со скидкой от количества занятий**

Покупатель выбирает `20` занятий по `1500` за каждое. Условное выражение проверяет минимальное количество для скидки, равное `10`. Двадцать проходит порог, поэтому в `discount` записывается `0.1`, то есть десять процентов в форме доли единицы.

Сначала количество и цена занятия дают полную стоимость `20 × 1500 = 30000`. В скобках `1 - discount` получается доля к оплате `0.9`. После умножения полной стоимости на эту долю печатается `27000.0`. Вещественная скидка делает вычисление итоговой цены вещественным.

При ровно `10` занятиях скидка тоже применяется из-за `>=`. При `9` занятиях выражение даёт целый `0`, множитель становится равным `1` и цена остаётся полной. В формулу подставляется доля `0.1`. Целое число `10` в той же формуле означало бы десять целых единиц, то есть тысячу процентов.

**Ожидаемый вывод**

```text
27000.0
```

````python
# Задаём объём покупки в 20 занятий.
lessons_count = 20
# Записываем цену одного занятия до применения скидки.
lesson_price = 1500

# Выбираем долю скидки 0.1 при покупке от десяти занятий включительно. Для 20 занятий это
# десять процентов.
discount = 0.1 if lessons_count >= 10 else 0

# Умножаем полную стоимость 30000 на долю к оплате 0.9. Скобки вычисляют эту долю до
# умножения.
total_price = lessons_count * lesson_price * (1 - discount)

# Печатаем стоимость с учётом скидки, равную 27000.0.
print(total_price)
````

## 2.2 Truthy(истинные) и falsy(ложные) значение;

**Пример 56. Проверка истинности чисел и строк с помощью bool()**

Чтобы увидеть, как значение воспринимается в условии, можно передать его функции `bool()`. Ноль среди обычных целых и вещественных чисел считается ложным, а ненулевые значения считаются истинными. Поэтому `10` и `-5` дают `True`, целый `0` даёт `False`, а в паре вещественных чисел `3.5` даёт `True` и `0.0` даёт `False`.

Для строки результат зависит от её длины. Строка `"Python"` содержит шесть символов, поэтому её логическое значение истинно. Пустая строка `""` содержит ноль символов и даёт `False`. Эти же правила использует `if`, когда строка стоит на месте условия.

Отрицательное число может быть истинным, как показано на `-5`, поскольку критерий для числа состоит в отличии от нуля. Строки `"0"`, `"False"` и `" "` считаются истинными, потому что содержат символы. Чтобы проверять содержимое или игнорировать пробелы, требуется отдельная операция.

**Ожидаемый вывод**

```text
True
True
False
True
False
True
False
```

````python
# Преобразуем положительное ненулевое целое 10 в True и печатаем результат.
print(bool(10))
# Проверяем отрицательное число −5. Оно отличается от нуля, поэтому даёт True.
print(bool(-5))
# Проверяем целочисленный ноль и получаем False.
print(bool(0))

# Проверяем ненулевое вещественное число 3.5 и получаем True.
print(bool(3.5))
# Проверяем вещественный ноль. Для 0.0 результат равен False.
print(bool(0.0))

# Проверяем строку из шести символов. Ненулевая длина даёт True.
print(bool("Python"))
# Проверяем пустую строку. Её длина равна нулю, поэтому печатается False.
print(bool(""))
````

**Пример 57. Строка и число в условиях if**

Выражением после `if` может быть сама переменная. Оператор интерпретирует её значение по тем же правилам, что и `bool()`. В первой проверке переменная `course_name` содержит строку `"Python"`. В ней шесть символов, поэтому выполняется ветвь с сообщением о наличии названия.

Во второй самостоятельной проверке переменная `students_count` содержит число `18`. Это ненулевое целое, поэтому выполняется ветвь с сообщением о наличии студентов. При пустом названии или нулевом количестве срабатывал бы `else` соответствующей проверки.

Прямая проверка особенно удобна, когда значения уже имеют ожидаемый смысл. Пробельная строка тоже истинна, потому что её длина положительна, а отрицательное количество было бы истинным как ненулевое число. При необходимости считать пробельное название пустым можно проверять `course_name.strip()`, а при необходимости проверять положительное количество подходит `students_count > 0`.

**Ожидаемый вывод**

```text
Название курса указано
На курсе есть студенты
```

````python
# Задаём непустое строковое название курса.
course_name = "Python"
# Сохраняем ненулевое целое количество студентов.
students_count = 18

# Проверяем истинность строки. Для «Python» она истинна, поскольку строка непуста.
if course_name:
    # Сообщаем о наличии названия в истинной ветви первой проверки.
    print("Название курса указано")
# Начинаем альтернативную ветвь первой проверки для пустого названия.
else:
    # Печатаем сообщение об отсутствии названия, когда проверяемая строка пуста.
    print("Название курса отсутствует")

# Начинаем отдельную проверку числа студентов. Для 18 условие истинно.
if students_count:
    # Печатаем сообщение о наличии студентов при ненулевом значении.
    print("На курсе есть студенты")
# Начинаем альтернативную ветвь проверки количества для случая, когда оно равно нулю.
else:
    # Сообщаем, что на курсе нет студентов, если указан ноль.
    print("На курсе нет студентов")
````

## 2.3 Логические операторы not, and, or, in, not in;

**Пример 58. Логическое отрицание с not**

Флаг `is_course_paid` описывает оплату и в данном запуске равен `False`. Оператор `not` проверяет истинность своего операнда и возвращает противоположное логическое значение. Выражение `not is_course_paid` читается как «обучение не оплачено» и даёт `True`.

При первом вычислении логическое значение выводится напрямую. Во второй раз то же выражение используется как условие `if`. Поскольку оно истинно, выполняется вложенная строка с напоминанием об оплате.

Если задать флагу `True`, выражение с `not` вернёт `False`, а тело `if` будет пропущено. Операция действует и на другие значения по правилам проверки истинности. Например, `not ""` даёт `True`, поскольку пустая строка считается ложной.

**Ожидаемый вывод**

```text
True
Необходимо оплатить обучение
```

````python
# Записываем False как логический признак отсутствия оплаты.
is_course_paid = False

# Инвертируем признак оплаты и печатаем True.
print(not is_course_paid)

# Повторяем отрицание в условии. При отсутствии оплаты оно истинно.
if not is_course_paid:
    # Выполняем тело if и печатаем напоминание об оплате.
    print("Необходимо оплатить обучение")
````

**Пример 59. Два обязательных условия с and**

Для выдачи сертификата требуется одновременно набрать хотя бы `60` баллов и оплатить обучение. Сравнение `exam_score >= 60` связывается сильнее `and`, поэтому выступает первым операндом логической операции. При балле `85` этот операнд даёт `True`.

Python вычисляет операнды `and` слева направо. Получив `True` слева, он читает значение `is_course_paid` справа. Оно тоже равно `True`, условие выполняется, и выводится сообщение о доступности сертификата.

Если балл ниже `60`, первый операнд ложен и Python пропускает чтение второго операнда. Такое поведение называется коротким замыканием. При проходном балле и ложном флаге оплаты второй операнд вычисляется и определяет ложный итог. В обоих случаях выполняется `else`.

**Ожидаемый вывод**

```text
Сертификат доступен
```

````python
# Сохраняем экзаменационный результат в 85 баллов, превышающий порог выдачи сертификата.
exam_score = 85
# Устанавливаем логический признак оплаты в True.
is_course_paid = True

# Сначала сравниваем экзаменационный балл с 60. При истинном результате проверяем оплату
# справа от and.
if exam_score >= 60 and is_course_paid:
    # Сообщаем о доступности сертификата, потому что необходимые условия выполнены
    # одновременно.
    print("Сертификат доступен")
# Обрабатываем все случаи, когда хотя бы одно из обязательных условий ложно.
else:
    # Печатаем общий отказ при недостаточном балле или отсутствии оплаты.
    print("Сертификат недоступен")
````

**Пример 60. Достаточное условие для скидки с or**

Скидка доступна при наличии промокода или при статусе постоянного студента. Для оператора `or` достаточно истинности хотя бы одного операнда. Флаг `has_promocode` равен `False`, поэтому Python переходит ко второму операнду.

Флаг `is_regular_student` равен `True`. Вычисление условия завершается истинным результатом, и выполняется строка о доступности скидки. Если бы оба флага имели значение `False`, выполнялась бы ветвь `else`.

Если первый операнд уже истинен, Python пропускает вычисление второго операнда. Например, при наличии промокода статус студента в этом условии не проверяется. При двух истинных признаках `or` тоже даёт истинный результат, поэтому условие подходит для скидки по любому из оснований.

**Ожидаемый вывод**

```text
Скидка доступна
```

````python
# Устанавливаем False, чтобы показать случай без промокода.
has_promocode = False
# Устанавливаем True для признака постоянного студента.
is_regular_student = True

# Пробуем левое основание скидки. Поскольку оно ложно, переходим к правому и получаем
# True.
if has_promocode or is_regular_student:
    # Печатаем сообщение о скидке, которую даёт статус постоянного студента.
    print("Скидка доступна")
# Выделяем ветвь для случая, когда оба основания скидки отсутствуют.
else:
    # Сообщаем, что при двух ложных признаках скидка недоступна.
    print("Скидка недоступна")
````

**Пример 61. Запасная строка для пустого названия**

Оператор `or` может возвращать значение одного из операндов. Сначала он вычисляет левый операнд. Если этот операнд истинен, его значение становится результатом всего выражения. Для ложного левого операнда Python вычисляет и возвращает правый.

В примере `course_name` содержит пустую строку, которая считается ложной. Поэтому выражение возвращает строку `"Курс без названия"`. Она присваивается имени `result` и затем печатается. При значении `"Python"` в `result` попало бы само название.

Этот приём выбирает запасное значение для любого ложного левого операнда. Он удобен, когда пустая строка должна обозначать отсутствие названия. Если пробелы тоже означают пустое название, строку сначала нужно обработать, поскольку строка `" "` считается истинной. При выборе запасного числового значения нужно учитывать, что `0` тоже считается ложным.

**Ожидаемый вывод**

```text
Курс без названия
```

````python
# Храним пустую строку как исходное название курса.
course_name = ""

# Проверяем истинность исходного названия и при его пустоте возвращаем правую строку
# целиком.
result = course_name or "Курс без названия"

# Печатаем строку, которую выбрал or. При текущем названии это «Курс без названия».
print(result)
````

**Пример 62. Поиск подстроки с in и not in**

При проверке `фрагмент in строка` Python ищет точную непрерывную последовательность символов. В названии `"Программирование на Python"` есть фрагмент `"Python"`, поэтому первая проверка даёт `True`. Фрагмента `"Java"` в нём нет, поэтому вторая проверка даёт `False`.

Оператор `not in` возвращает противоположный результат той же проверки вхождения. Поэтому `"Python" not in course_name` даёт `False`, а `"Java" not in course_name` даёт `True`. Каждый вызов `print()` выводит одно логическое значение.

Поиск чувствителен к регистру, так что строчное `"python"` в текущем названии отсутствует. Оператор проверяет подстроку без требования к границам слов. Например, `"Py"` входит в название как начало слова `Python`. Пустая строка считается подстрокой любой строки.

**Ожидаемый вывод**

```text
True
False
False
True
```

````python
# Записываем полное название курса, в конце которого находится слово Python.
course_name = "Программирование на Python"

# Ищем непрерывную подстроку Python и печатаем True.
print("Python" in course_name)
# Ищем подстроку Java, которой нет в названии, и печатаем False.
print("Java" in course_name)

# Проверяем отсутствие Python. Этот фрагмент найден, поэтому получаем False.
print("Python" not in course_name)
# Проверяем отсутствие Java. Такого фрагмента нет, поэтому получаем True.
print("Java" not in course_name)
````

**Пример 63. Приоритет and и or и влияние скобок**

У `and` приоритет выше, чем у `or`. Первая строка разбирается как `True or (False and False)`. Внешний оператор здесь `or`, поэтому первым вычисляется левый операнд `True`. Он уже определяет истинный результат, и Python пропускает всю правую часть. Печатается `True`.

Во второй строке скобки меняют группировку на `(True or False) and False`. Теперь внешний оператор `and`. Сначала выражение в скобках даёт `True`, причём внутренний `or` пропускает свой правый операнд. Затем внешний `and` вычисляет правый операнд `False`, который становится общим результатом.

Приоритет определяет структуру выражения. Порядок вычисления операндов и короткое замыкание определяют, какие части этой структуры фактически выполняются. Для логических операторов общий порядок связывания от более сильного к более слабому выглядит так `not`, затем `and`, затем `or`. Скобки позволяют явно задать нужную группировку.

**Ожидаемый вывод**

```text
True
False
```

````python
# Группируем правые операнды по приоритету and. Левый True позволяет оператору or вернуть
# True и пропустить всю правую часть.
print(True or False and False)
# Сначала выражение в скобках даёт True. Внешний and вычисляет правый False и возвращает
# его.
print((True or False) and False)
````

**Пример 65. Доступ к отдельным символам по индексу**

Функция `type(course_code[0])` показывает, что результат индексации имеет тип `str`. В Python один символ представлен строкой длины `1`. Квадратные скобки с целым индексом выполняют чтение элемента, поэтому исходное значение `course_code` сохраняется. **Пример 66. Отрицательные индексы и отсчёт справа**

Отрицательный индекс выбирает позицию с правого края строки. Индекс `-1` обозначает последний символ, `-2` обозначает предпоследний, а `-3` обозначает третий справа. В `PYTHON` этим позициям соответствуют `N`, `O` и `H`.

Для допустимого отрицательного индекса можно получить эквивалентную позицию слева, прибавив длину строки. Здесь длина равна `6`, поэтому `-1` переводится в `5`, `-2` в `4`, `-3` в `3`, а `-6` в `0`. Последний вызов поэтому выбирает начальную букву `P`.

Отсчёт справа удобен, когда длина строки меняется, а программа всегда читает последний символ. Позиции для этой строки лежат от `-6` до `-1` при отсчёте справа и от `0` до `5` при отсчёте слева. Индекс `-7` выходит за левый край, а пустая строка не имеет позиции `-1`. В обоих случаях доступ по индексу вызывает `IndexError`.

**Ожидаемый вывод**

```text
N
O
H
P
```

````python
# Сохраняем строку PYTHON, у которой шесть позиций для отсчёта с каждого края.
course_code = "PYTHON"

# Индекс -1 выбирает крайний правый символ N.
print(course_code[-1])
# Индекс -2 сдвигается на одну позицию влево и выбирает O.
print(course_code[-2])
# Индекс -3 выбирает третий справа символ H.
print(course_code[-3])
# Индекс -6 отсчитывает все шесть позиций и выбирает самую левую букву P.
print(course_code[-6])
````

Для получения последнего символа в этой строке подойдёт `course_code[5]` или `course_code[-1]`. Если позиция `6` нужна как часть алгоритма, длину строки нужно сначала проверить.
````python
# Сохраняем строку длиной шесть. При отсчёте слева доступны индексы от 0 до 5
# включительно.
course_code = "PYTHON"

# Запрашиваем седьмой символ по индексу 6. Python выбрасывает IndexError при вычислении
# аргумента, поэтому print не вызывается.
print(course_code[6])
````

**Пример 68. Проверка длины перед чтением третьего символа**

Программа хочет прочитать символ с индексом `2`. Это третья позиция, поэтому строка должна содержать как минимум три символа. Условие `len(course_code) >= 3` устанавливает именно эту границу и допускает строку длиной ровно три.

В переменной находится `"PY"`. Функция `len` возвращает `2`, выражение `2 >= 3` даёт `False`, и интерпретатор сразу переходит к телу `else`. На экран выводится сообщение `Недостаточно символов`. Выражение `course_code[2]` расположено внутри ветки `if`, поэтому в этом проходе оно вообще не вычисляется.

Если подставить `"PYT"`, длина станет равной `3` и ветка `if` выведет `T`. Та же проверка сработает для любой более длинной строки. Условие `len(course_code) >= 2` было бы ошибочным. Оно допустило бы исходное значение `"PY"`, у которого доступен индекс `1` и отсутствует индекс `2`.

**Ожидаемый вывод**

```text
Недостаточно символов
```

````python
# Задаём строку из двух букв, чтобы показать случай, когда третьей позиции нет.
course_code = "PY"

# Проверяем, что длины хватает для чтения символа с индексом 2.
if len(course_code) >= 3:
    # Читаем и печатаем третий символ только после успешной проверки длины.
    print(course_code[2])
# Переходим к запасной ветке, если длина оказалась меньше трёх.
else:
    # Сообщаем о недостаточной длине и завершаем обработку короткой строки.
    print("Недостаточно символов")
````

**Пример 69. Срез строки с заданными началом и концом**

Срез `course_code[start:stop]` возвращает строку из последовательных символов, начиная с позиции `start`. Позиция `stop` в результат не входит. В `PYTHON-2026` буквы занимают позиции от `0` до `5`, дефис стоит на позиции `6`, а цифры года занимают позиции от `7` до `10`.

В `course_code[0:6]` выбираются индексы `0`, `1`, `2`, `3`, `4` и `5`, поэтому получается `PYTHON`. Дефис на позиции `6` остаётся за правой границей. Срез `course_code[7:11]` берёт индексы от `7` до `10` и даёт `2026`. Стоп-граница `11` равна длине всей строки и позволяет включить последний символ.

Срез `course_code[1:4]` начинает чтение с `Y` и заканчивает на `H`, поэтому даёт `YTH`. Здесь выбраны три позиции, что совпадает с разностью `4 - 1`. Такой расчёт длины удобен для обеих заданных неотрицательных границ внутри строки при шаге `1`. Чтобы включить символ с индексом `4` в этот результат, нужно передать стоп-границу `5`.

**Ожидаемый вывод**

```text
PYTHON
2026
YTH
```

````python
# Задаём код с шестибуквенным названием, дефисом и годом.
course_code = "PYTHON-2026"

# Берём позиции от 0 до 5. Стоп-граница 6 исключает дефис, поэтому срез даёт PYTHON.
print(course_code[0:6])
# Берём позиции от 7 до 10. Начинаем сразу после дефиса и получаем 2026.
print(course_code[7:11])
# Берём позиции 1, 2 и 3. Начальная граница входит в срез, поэтому результат равен YTH.
print(course_code[1:4])
````

**Пример 70. Срезы с опущенными границами**

При обычном движении слева направо Python может подставить границы среза автоматически. Если опущено начало, чтение начинается с индекса `0`. Если опущен конец, чтение продолжается до конца строки. При длине `11` это соответствует стоп-границе `11`.

В `course_code[:6]` начало опущено, поэтому срез выбирает шесть первых позиций и даёт `PYTHON`. В `course_code[7:]` задана начальная позиция `7`. На ней стоит первая цифра года, поэтому срез включает все четыре цифры `2026`.

В срезе `course_code[:]` опущены обе границы. В вывод попадает всё содержимое `PYTHON-2026`. Например, `[:6]` сохраняет первые шесть символов даже при более длинном продолжении строки, а длина `[7:]` зависит от того, сколько символов осталось после первых семи позиций. Для чтения одного символа с позиции `7` используется индекс `[7]`.

**Ожидаемый вывод**

```text
PYTHON
2026
PYTHON-2026
```

````python
# Сохраняем составной код, чтобы получить его начало, конец и полную строку.
course_code = "PYTHON-2026"

# Пропускаем левую границу и берём символы с начала до позиции 6 без её включения.
print(course_code[:6])
# Начинаем с позиции 7 и читаем оставшиеся четыре символа года.
print(course_code[7:])
# Опускаем обе границы. Срез будет содержать строку целиком.
print(course_code[:])
````

**Пример 71. Отрицательные границы строкового среза**

Отрицательные границы среза отсчитываются от правого края. В строке `PYTHON-2026` длиной `11` значение `-4` соответствует позиции `7`, `-5` соответствует позиции `6`, а `-2` соответствует позиции `9`. После определения позиций срез с обычным шагом `1` движется слева направо.

Срез `course_code[-4:]` начинается на первом символе года и возвращает `2026`. В `course_code[:-5]` начало опущено, а исключаемая правая граница `-5` указывает на дефис. Поэтому результат содержит шесть букв `PYTHON`.

В `course_code[-4:-2]` выбираются позиции `7` и `8`, что даёт `20`. Правая граница `-2` указывает на позицию `9`, и находящаяся там цифра `2` уже не входит. Такой отсчёт удобен для суффикса фиксированной длины. Если строка короче четырёх символов, `[-4:]` вернёт её целиком, поэтому при проверке формата нужно отдельно убедиться в нужной длине.

**Ожидаемый вывод**

```text
2026
PYTHON
20
```

````python
# Записываем строку с четырёхсимвольным годом после дефиса.
course_code = "PYTHON-2026"

# Начинаем с четвёртого элемента справа и читаем до конца. Получается 2026.
print(course_code[-4:])
# Читаем с начала до пятого элемента справа. Этот элемент является дефисом и в срез не
# входит.
print(course_code[:-5])
# Берём позиции 7 и 8 между двумя границами, отсчитанными справа. Получается 20.
print(course_code[-4:-2])
````

**Пример 72. Выбор символов с положительным шагом**

Третье значение в выражении `numbers[start:stop:step]` задаёт расстояние между соседними выбранными позициями. При положительном шаге срез идёт слева направо. Строка `"0123456789"` удобна для наблюдения, поскольку на каждой позиции написана её цифра.

Срез `[0:10:2]` начинает с позиции `0`, затем выбирает `2`, `4`, `6` и `8`. Следующей была бы позиция `10`, которая совпадает с исключаемой границей. Результат равен `02468`. При тех же конце и шаге срез `[1:10:2]` начинается с `1` и выбирает `1`, `3`, `5`, `7` и `9`, образуя `13579`.

В срезе `[::3]` стандартные границы охватывают всю строку, а шаг `3` даёт позиции `0`, `3`, `6` и `9`. Получается `0369`. Срез `[3::3]` начинает сразу с `3`, поэтому даёт `369`. Шаг `3` оставляет два пропущенных символа между выбранными. Нулевой шаг недопустим и приводит к `ValueError`.

**Ожидаемый вывод**

```text
02468
13579
0369
369
```

````python
# Используем цифры в строке, чтобы значение каждого символа подсказывало его индекс.
numbers = "0123456789"

# Начинаем с нуля и прибавляем два к каждой выбранной позиции. Последняя выбранная позиция
# равна 8.
print(numbers[0:10:2])
# Начинаем с единицы и сохраняем шаг два. Выбираем пять символов на нечётных позициях.
print(numbers[1:10:2])
# Идём от начала до конца с шагом три. Печатаем символы на позициях 0, 3, 6 и 9.
print(numbers[::3])
# Начинаем с позиции 3 и движемся тем же шагом до конца, поэтому первый ноль из
# предыдущего результата пропускается.
print(numbers[3::3])
````

**Пример 73. Обратный срез и движение справа налево**

Отрицательный шаг меняет направление среза. При опущенных границах чтение начинается с последнего символа и продолжается в сторону первого, пока выбранная позиция входит в строку. В `PYTHON` индексы при таком обходе убывают от `5` к `0`.

Срез `course_code[::-1]` посещает позиции `5`, `4`, `3`, `2`, `1` и `0`, так что результатом становится `NOHTYP`. Шаг `-2` уменьшает индекс сразу на два. Поэтому `course_code[::-2]` берёт `5`, `3` и `1`, образуя `NHY`.

В `course_code[4:1:-1]` чтение начинается на позиции `4`, где стоит `O`, затем выбираются `H` на позиции `3` и `T` на позиции `2`. Индекс `1` остаётся за стоп-границей, поэтому получается `OHT`. Для отрицательного шага эффективная начальная позиция должна находиться правее стоп-границы. Например, `[1:4:-1]` даёт пустой срез.

**Ожидаемый вывод**

```text
NOHTYP
NHY
OHT
```

````python
# Сохраняем строку PYTHON для чтения справа налево.
course_code = "PYTHON"

# Шаг -1 выбирает все символы от последнего до первого и разворачивает строку.
print(course_code[::-1])
# Шаг -2 берёт буквы на позициях 5, 3 и 1. Между двумя выбранными символами пропускается
# один.
print(course_code[::-2])
# Идём от позиции 4 к позиции 1 и останавливаемся перед ней. В срез входят O, H и T.
print(course_code[4:1:-1])
````

**Пример 74. Границы среза за пределами строки**

При срезе с положительным шагом Python подстраивает числовые границы под доступные позиции. Стоп-граница `100` в `course_code[0:100]` ограничивается длиной строки `6`. В результате читаются все шесть символов и получается `PYTHON`.

В `course_code[-100:3]` отрицательная начальная граница уходит дальше начала строки. Python сводит её к `0`, а конец `3` остаётся на месте. Выбираются индексы `0`, `1` и `2`, поэтому получается `PYT`.

В срезе `course_code[10:]` эффективное начало оказывается в конце строки, поэтому выбирать уже нечего. В `course_code[4:2]` начало правее конца, а обычный шаг движется вправо, поэтому этот срез тоже пуст. `print` печатает для каждой пустой строки только перевод строки. У индексации правила строже. Например, `course_code[10]` сразу вызывает `IndexError`, хотя срез `[10:]` допустим.

````python
# Задаём строку длиной шесть для примеров с границами за её краями.
course_code = "PYTHON"

# Стоп-граница 100 сокращается до длины строки. В срез входят все буквы.
print(course_code[0:100])
# Левая граница -100 смещается к началу строки. Индекс 3 исключён, поэтому получается PYT.
print(course_code[-100:3])
# Начало выходит за правый край. После ограничения границы срез оказывается пустым.
print(course_code[10:])
# Шаг по умолчанию равен 1, а граница 2 находится левее начальной позиции 4. Срез
# получается пустым.
print(course_code[4:2])
````

**Пример 75. Повторное использование границ с объектами slice**

Функция `slice` создаёт объект, который хранит параметры среза. В `name_slice` сохраняются начало `0` и конец `6`. Выражение `first_code[name_slice]` применяет эти границы так же, как запись `first_code[0:6]`, поэтому возвращает первые шесть символов `PYTHON`. Для второго кода те же границы выделяют `GOLANG`.

В `slice(-4, None)` начальная граница отсчитывается от конца, а `None` обозначает опущенную стоп-границу. Запись `first_code[year_slice]` соответствует `first_code[-4:]` и даёт `2026`. Когда тот же объект применяется к `second_code`, возвращается `2025`. Символы выбираются при обращении к конкретной строке.

Объект среза удобно хранить отдельно, если одинаковые позиции имеют общий смысл для нескольких строк. Здесь выбор имени по фиксированным шести позициям предполагает, что обе части до дефиса имеют одинаковую длину. При более коротком названии такой срез может захватить дефис или начало года. Срез последних четырёх символов предполагает, что год действительно стоит в конце кода.

**Ожидаемый вывод**

```text
PYTHON
GOLANG
2026
2025
```

````python
# Сохраняем первый код, в котором название PYTHON занимает шесть позиций.
first_code = "PYTHON-2026"
# Сохраняем второй код с шестибуквенным названием GOLANG и другим годом.
second_code = "GOLANG-2025"

# Создаём объект границ для первых шести символов. Стоп-граница 6 исключается.
name_slice = slice(0, 6)
# Создаём объект границ для последних четырёх символов. None обозначает чтение до конца.
year_slice = slice(-4, None)

# Применяем срез названия к первому коду и получаем PYTHON.
print(first_code[name_slice])
# Применяем тот же объект к строке второго кода и получаем GOLANG.
print(second_code[name_slice])

# Применяем срез года к первому коду и получаем 2026.
print(first_code[year_slice])
# Повторно используем срез года со второй строкой и получаем 2025.
print(second_code[year_slice])
````

**Пример 76. Цикл while со счётчиком занятий**

`lesson_number` начинается с `1` и хранит номер занятия для текущего прохода. `while` проверяет условие перед каждым проходом. Пока `lesson_number <= 5` истинно, выполняются обе строки с отступом. Функция `print` вставляет текущий номер в фигурные скобки f-строки, а `lesson_number += 1` увеличивает счётчик на единицу.

При первом проходе счётчик равен `1`, поэтому печатается `Занятие №1`, после чего счётчик становится равным `2`. Такая же пара действий повторяется для `2`, `3`, `4` и `5`. Условие включает число `5`, так что пятое занятие тоже выводится. Затем счётчик становится равным `6`, и следующая проверка `6 <= 5` даёт `False`.

Строка `Занятия завершены` печатается один раз после цикла. У неё нет отступа тела `while`. Если задать счётчику начальное значение `6`, тело пропустится целиком, а итоговая строка всё равно появится. Если убрать увеличение счётчика при начальном значении `1`, условие будет истинным при каждой проверке.

**Ожидаемый вывод**

```text
Занятие №1
Занятие №2
Занятие №3
Занятие №4
Занятие №5
Занятия завершены
```

````python
# Устанавливаем счётчик на номер первого занятия.
lesson_number = 1

# Перед каждым проходом проверяем, что текущий номер не превышает 5.
while lesson_number <= 5:
    # Подставляем текущий номер в f-строку и печатаем его.
    print(f"Занятие №{lesson_number}")
    # Увеличиваем номер на единицу. После пятого занятия счётчик станет равным 6, что
    # завершит цикл на следующей проверке.
    lesson_number += 1

# Печатаем итог после выхода из тела цикла. Отступа цикла у этой строки нет.
print("Занятия завершены")
````

**Пример 77. Накопление общей длительности занятий**

Счётчик `lesson_number` определяет номер текущего занятия, а накопитель `total_hours` содержит сумму длительностей уже обработанных занятий. До начала цикла номер равен `1`, накопленная сумма равна `0`, а длительность каждого занятия задана как `1.5` часа.

Условие `lesson_number <= 5` позволяет обработать пять занятий. В начале каждого прохода к накопителю прибавляется `1.5`. Первый проход выводит номер `1` и сумму `1.5`, второй выводит номер `2` и сумму `3.0`. Следующие значения равны `4.5`, `6.0` и `7.5`. После первого сложения с `1.5` значение накопителя имеет тип `float`, поэтому целые по величине суммы в выводе выглядят как `3.0` и `6.0`.

Счётчик увеличивается после двух строк вывода, поэтому номер и накопленная сумма относятся к одному занятию. После вывода номера `5` счётчик становится равным `6`, и условие завершает цикл. Итоговая строка берёт накопленное значение `7.5`. Если присваивать `total_hours = 0` внутри тела, каждый проход будет сбрасывать сумму и показывать только длительность текущего занятия.

**Ожидаемый вывод**

```text

Общая продолжительность: 1.5 ч.
Занятие №2
Общая продолжительность: 3.0 ч.
Занятие №3
Общая продолжительность: 4.5 ч.
Занятие №4
Общая продолжительность: 6.0 ч.

Итого: 7.5 ч.
```

````python
# Задаём номер первого занятия, с которого начнётся обработка.
lesson_number = 1
# Храним длительность одного занятия в часах. Значение 1.5 имеет тип float.
lesson_duration = 1.5
# Инициализируем накопитель нулём перед циклом, чтобы сумма сохранялась между проходами.
total_hours = 0

# Входим в тело, если номер текущего занятия не превышает пяти.
while lesson_number <= 5:
    # Прибавляем длительность текущего занятия к ранее накопленной сумме.
    total_hours += lesson_duration

    # Печатаем номер занятия до изменения счётчика.
    print(f"Занятие №{lesson_number}")
    # Печатаем сумму длительностей по текущему занятию включительно.
    print(f"Общая продолжительность: {total_hours} ч.")

    # Переходим к следующему номеру. После номера 5 следующая проверка остановит цикл.
    lesson_number += 1

# После цикла печатаем общую длительность всех занятий, равную 7.5 часа.
print(f"Итого: {total_hours} ч.")
````

**Пример 78. Цикл while до достижения минимального порога**

В начале в группе `5` студентов, а для формирования группы требуется `20`. Условие `students_count < required_students` проверяется до каждой очередной прибавки. Если студентов меньше порога, тело добавляет сразу трёх студентов и выводит новое значение.

После проходов в группе последовательно становится `8`, `11`, `14`, `17` и `20` студентов. Начальное значение `5` не печатается, поскольку вызов `print` стоит после прибавки. Когда счётчик достигает `20`, сравнение `20 < 20` даёт `False`. Следующей прибавки нет, и после цикла печатается `Группа сформирована`.

Правило означает «продолжать, пока студентов меньше порога». Если начать с `6`, серия значений достигнет `21`, а условие всё равно остановит цикл. При начальном количестве `20` или больше тело пропускается. Условие `students_count != required_students` при старте с `6` было бы ошибочным. Прибавление по три никогда не дало бы ровно `20`, и остановки не произошло бы.

Тело цикла состоит из одного вызова `print(symbol)`. По умолчанию после каждого вызова `print` добавляет перевод строки, поэтому шесть букв появляются на шести отдельных строках. После выдачи последней буквы в последовательности больше нет элементов, и цикл завершается.

Здесь программе нужен сам символ, поэтому прямой перебор передаёт его телу автоматически. Если позже понадобится одновременно знать порядковый индекс, можно перебрать `enumerate(course_code)`, которое даёт пары из индекса и символа. Если исходная строка пуста, `for` не выполнит тело ни разу.

**Пример 81. Подсчёт цифр при переборе строки**

Цикл обходит всю строку `PYTHON-2026`, а целочисленный счётчик `digits_count` начинается с `0`. На каждом проходе условие `symbol in "0123456789"` проверяет, входит ли текущий односимвольный элемент в строку из десяти цифр от `0` до `9`.

Для букв `P`, `Y`, `T`, `H`, `O`, `N` и дефиса условие ложно, поэтому счётчик остаётся равным `0`. Затем рассматриваются `2`, `0`, `2` и `6`. Все четыре проверки истинны. Счётчик становится равным `1`, `2`, `3` и `4`, причём повторное появление символа `2` учитывается как отдельная позиция.

После окончания обхода программа один раз печатает `Количество цифр: 4`. При строке без подходящих символов будет выведен ноль. Условие специально перечисляет цифры `0`–`9`. Символы других систем письма, например `٣`, оно пропустит. Помещать строку итогового вывода внутрь `for` имеет смысл только для показа промежуточных сумм на каждом проходе.

**Ожидаемый вывод**

```text
Количество цифр: 4
```

````python
# Задаём строку, содержащую буквы, дефис и четыре цифры.
course_code = "PYTHON-2026"
# Обнуляем счётчик до начала обхода, чтобы считать каждое совпадение в этой строке.
digits_count = 0

# Передаём телу по одному символу слева направо, пока строка не закончится.
for symbol in course_code:
    # Проверяем, входит ли текущий символ в строку из цифр от 0 до 9.
    if symbol in "0123456789":
        # Добавляем единицу только при попадании символа в указанный набор цифр.
        digits_count += 1

# После полного обхода выводим число найденных позиций с цифрами, равное 4.
print(f"Количество цифр: {digits_count}")
````

`range(5)` создаёт компактное представление последовательности целых чисел. Один аргумент задаёт исключаемую границу `5`, а начало и шаг по умолчанию равны `0` и `1`. Поэтому `numbers` представляет пять значений `0`, `1`, `2`, `3` и `4`.

После трёх диагностических строк цикл присваивает `number` каждое значение от `0` до `4` и печатает его отдельно. `5` не появляется, поскольку это стоп-граница. Если нужны числа от `1` до `5` включительно, соответствующий диапазон задаётся как `range(1, 6)`. В отличие от отдельного итератора, тот же объект `range` можно снова передать в другой `for` и получить все его значения с начала.

Внешний `for` получает номера групп `1`, `2` и `3` из `range(1, 4)`. Для каждого номера выполнение доходит до внутреннего `for`, который получает номера занятий `1` и `2` из `range(1, 3)`. Пока внутренний цикл выполняется, значение `group` сохраняется.

Сначала выводятся занятия `1` и `2` группы `1`. Внутренний диапазон заканчивается, и внешний цикл получает следующую группу. Всего выполняются `3 * 2 = 6` вызовов `print` в порядке пар `(1, 1)`, `(1, 2)`, `(2, 1)`, `(2, 2)`, `(3, 1)`, `(3, 2)`.

Вызов `print` расположен с отступом внутри обоих циклов, поэтому печатает каждую пару. Если уменьшить его отступ на один уровень и поместить после внутреннего цикла в тело внешнего, для каждой группы появится одна строка с последним присвоенным номером занятия `2`.

**Пример 86. Сброс внутреннего счётчика во вложенных циклах while**

Внешний счётчик `group` начинается с `1` и управляет проходами по трём группам. В начале каждого внешнего прохода `lesson` получает `1`. Внутренний `while` печатает текущую пару, увеличивает `lesson` и повторяет действия, пока номер занятия не превысит `2`.

В группе `1` внутренний счётчик принимает значения `1` и `2` при печати. После второго вывода он становится равным `3`, и внутренний цикл заканчивается. Только после этого `group` увеличивается до `2`. Для групп `2` и `3` происходит та же последовательность. После группы `3` значение `group` становится равным `4`, поэтому внешнее условие ложно. Напечатано шесть пар.

Размещение `lesson = 1` внутри внешнего тела даёт каждой группе собственный проход по занятиям. Если перенести присваивание перед внешним циклом, после первой группы `lesson` останется равным `3`, и для групп `2` и `3` внутреннее тело пропустится. У каждого `while` есть своё изменение счётчика в соответствующем теле, которое позволяет дойти до ложного условия.

**Ожидаемый вывод**

```text
Группа 1, занятие 1
Группа 1, занятие 2
Группа 2, занятие 1
Группа 2, занятие 2
Группа 3, занятие 1
Группа 3, занятие 2
```

````python
# Инициализируем счётчик групп до входа во внешний цикл.
group = 1

# Обрабатываем группы, пока счётчик group не превышает 3.
while group <= 3:
    # Перед занятиями каждой группы сбрасываем внутренний счётчик к 1.
    lesson = 1

    # Продолжаем занятия текущей группы, пока lesson не превышает 2.
    while lesson <= 2:
        # Выводим текущие значения обоих счётчиков как пару группы и занятия.
        print(f"Группа {group}, занятие {lesson}")
        # Увеличиваем номер занятия. После 2 значение 3 завершит внутренний цикл.
        lesson += 1

    # После завершения внутреннего цикла увеличиваем номер группы. После группы 3 внешний
    # цикл встретит значение 4 и завершится.
    group += 1
````

Внешний цикл перебирает пять строк с номерами от `1` до `5`. На каждой строке внутренний цикл использует `range(1, row + 1)`. Вызов `print()` с отступом внешнего тела выполняется один раз после всех звёздочек очередной строки и добавляет перевод на следующую строку. Если оставить стандартное окончание у внутреннего `print`, каждая звёздочка появится на отдельной строке. Если задать `range(1, row)`, каждый ряд потеряет одну звёздочку, а первый окажется пустым.

**Пример 88. Значение None, его тип и преобразование к bool**

`None` здесь обозначает отсутствие результата экзамена. Присваивание `exam_score = None` создаёт переменную с заданным значением, и Python позволяет читать её, передавать в функции и проверять в условиях. Кавычки здесь отсутствуют. Если записать `"None"`, получится обычная строка из четырёх букв.

Первый вызов `print` показывает `None`, потому что так выглядит его текстовое представление. Второй выводит `<class 'NoneType'>`, поскольку единственное значение `None` относится к типу `NoneType`. Третий передаёт значение функции `bool` и печатает `False`.

Признак ложности используется в условиях. При `if exam_score:` тело пропустилось бы, если оценка равна `None`. Тело также пропустилось бы при оценке `0`, поскольку ноль тоже преобразуется к `False`. Когда нулевая оценка допустима, наличие результата можно проверять условием `exam_score is not None`.

**Ожидаемый вывод**

```text
None

```

У числа `0` собственное значение и тип `int`, поэтому проверка даёт `True`. Python выполняет тело `if`, f-строка подставляет `0`, и на экране появляется `Полученный результат: 0`. Условие `if exam_score:` проверяло бы логическое значение оценки. Для `0` оно ложно, и такая проверка потеряла бы полученный результат. Выражение `is not None` отвечает на конкретный вопрос о наличии значения. **Пример 90. Поиск первой цифры с помощью None и break**

`found_symbol` до обхода равно `None` и обозначает, что подходящий символ пока не найден. Цикл читает `PYTHON-2026` слева направо. Условие `symbol in "0123456789"` пропускает шесть букв и дефис, потому что их нет в наборе цифр.

На позиции `7` цикл встречает строковый символ `"2"`. Условие становится истинным, и значение сохраняется в `found_symbol`. Сразу после присваивания `break` завершает цикл. Символы `0`, `2` и `6` дальше по строке не проверяются, поэтому переменная хранит первую встреченную цифру. При отсутствии `break` последующие совпадения перезаписали бы её, и для этой строки сохранилось бы `"6"`.

После цикла условие `found_symbol is not None` истинно, и ячейка печатает `Первая найденная цифра: 2`. Если строка будет пустой или не будет содержать ни одной из цифр `0`–`9`, `found_symbol` сохранит значение `None`. Тогда выполнится ветка `else` с сообщением `Цифры отсутствуют`. Найденный символ остаётся строкой длиной один символ.

**Ожидаемый вывод**

```text
Первая найденная цифра: 2
```

````python
# Задаём последовательность, в которой будем искать первую цифру слева.
course_code = "PYTHON-2026"
# До начала поиска обозначаем отсутствие совпадения значением None.
found_symbol = None

# Перебираем символы слева направо, по одному на каждой итерации.
for symbol in course_code:
    # Проверяем попадание текущего символа в набор цифр от 0 до 9.
    if symbol in "0123456789":
        # Запоминаем первый подходящий символ. Для этой строки он равен строке из одной
        # цифры 2.
        found_symbol = symbol
        # Завершаем цикл сразу после сохранения совпадения, чтобы последующие цифры его не
        # заменили.
        break

# После цикла проверяем, было ли заменено исходное значение None.
if found_symbol is not None:
    # Печатаем символ, если поиск завершился с совпадением.
    print(f"Первая найденная цифра: {found_symbol}")
# Выбираем запасную ветку, если найденное значение осталось равным None.
else:
    # Сообщаем, что в строке отсутствуют цифры из проверяемого набора.
    print("Цифры отсутствуют")
````

## 3.4 Оператор break и continue. Блок else у циклов while и for;

**Пример 91. Прерывание цикла командой break**

`range(1, 10)` может передать циклу числа от `1` до `9`. Тело сначала проверяет `number == 5` и только затем может вывести текущее значение. Для `1`, `2`, `3` и `4` сравнение ложно, поэтому эти четыре числа печатаются.

Когда цикл получает `5`, сравнение становится истинным и выполняется `break`. Управление сразу покидает этот цикл. Вызов `print(number)` для текущего числа пропускается, а значения `6`, `7`, `8` и `9` уже не запрашиваются из диапазона.

После прерывания выполняется первая строка за пределами цикла, которая печатает `Цикл завершён`. Положение проверки определяет, появится ли число `5` на экране. Если перенести печать перед условием, это число успеет появиться до прерывания. Во вложенном коде `break` завершает ближайший охватывающий цикл.

**Ожидаемый вывод**

```text
1
2
3
4
Цикл завершён
```

````python
# Готовим числа от 1 до 9 для перебора. Выход из тела возможен раньше исчерпания
# диапазона.
for number in range(1, 10):
    # Проверяем, достиг ли текущий элемент значения, на котором нужно остановиться.
    if number == 5:
        # При значении 5 сразу выходим из ближайшего цикла, пропуская остальное его тело.
        break

    # Печатаем число, если проверка не прервала цикл. В этом примере выводятся числа от 1
    # до 4.
    print(number)

# Продолжаем выполнение после цикла и один раз печатаем итоговое сообщение.
print("Цикл завершён")
````

**Пример 92. Пропуск чётных чисел командой continue**

`for` получает числа `1`–`10` из `range(1, 11)`. Для каждого числа оператор `%` вычисляет остаток от деления на два. У чётного числа остаток равен нулю, поэтому `number % 2 == 0` истинно для `2`, `4`, `6`, `8` и `10`.

На чётных значениях `continue` завершает текущий проход сразу. `print(number)` находится ниже и для этих чисел не выполняется. После `continue` цикл `for` получает следующий элемент диапазона самостоятельно. На нечётных значениях условие ложно, и число доходит до печати.

На экране появляются `1`, `3`, `5`, `7` и `9`. Значение `10` тоже было обработано условием, после чего `continue` запросил очередной проход. В диапазоне больше нет чисел, поэтому цикл завершился. Если перенести печать выше проверки, на экране будут и чётные числа, поскольку `continue` пропускает только оставшиеся команды текущего прохода.

**Ожидаемый вывод**

```text
1
3
5
7
9
```

````python
# Передаём циклу числа от 1 до 10 включительно для проверки чётности.
for number in range(1, 11):
    # Вычисляем остаток от деления текущего числа на 2. Нулевой остаток означает чётное
    # число.
    if number % 2 == 0:
        # Пропускаем оставшееся тело на текущем чётном числе и запрашиваем следующий
        # элемент.
        continue

    # Печатаем текущее число, если до этой строки выполнение дошло. Так в вывод попадают
    # только нечётные значения.
    print(number)
````

**Пример 93. Пропуск чётных чисел с обновлением счётчика в while**

Перед циклом `number` равно `0`. На входе в каждом проходе проверяется `number < 10`, затем счётчик сразу увеличивается на единицу. Поэтому проверка чётности в теле первый раз получает `1`, а последний раз получает `10`.

Для нечётных значений `1`, `3`, `5`, `7` и `9` условие чётности ложно, и число печатается. Для `2`, `4`, `6`, `8` и `10` срабатывает `continue`, после которого выполнение возвращается к условию `while`. Когда уже обработано число `10`, проверка `10 < 10` завершает цикл.

Увеличение счётчика нужно выполнить до потенциального `continue`. Если перенести `number += 1` в самый низ, начальное чётное значение `0` дойдёт до `continue` и будет пропускать увеличение. Тогда при каждой новой проверке получится `0 < 10`, и цикл не завершится. Размещение прибавки в начале гарантирует изменение значения на каждом проходе.

**Ожидаемый вывод**

```text
1
3
5
7
9
```

````python
# Задаём число 0, чтобы первое увеличение в теле дало 1.
number = 0

# Проверяем текущее значение до прибавки. Если оно достигло 10, тело больше не
# запускается.
while number < 10:
    # Прибавляем единицу заранее, чтобы счётчик вырос даже перед проходом, который
    # завершит continue.
    number += 1

    # Смотрим, делится ли уже увеличенное число на два без остатка.
    if number % 2 == 0:
        # При чётном числе возвращаемся к проверке while, пропуская печать текущего
        # значения.
        continue

    # Печатаем нечётное число. Для чётного значения управление до этой строки не доходит.
    print(number)
````

**Пример 94. Выполнение else после полного обхода без совпадения**

Цикл ищет заглавную букву `A` в строке `PYTHON`. На каждом проходе он сравнивает очередной символ с `search_symbol`. Сравнения для `P`, `Y`, `T`, `H`, `O` и `N` дают `False`, поэтому условия для сообщения о находке и прерывания ни разу не выполняются.

После символа `N` строка исчерпана, и внешняя конструкция `for` доходит до своего `else`. В этой ветке один раз печатается `Символ отсутствует`. У `else` тот же отступ, что у `for`, поэтому между двумя конструкциями есть связь. Вложенный `if` имеет собственный дополнительный отступ.

`else` цикла выполняется при нормальном исчерпании элементов без `break`. Пустая строка тоже приведёт к этой ветке, хотя тело `for` не выполнится ни разу. Если бы условие совпадения сработало и тело дошло до `break`, Python вышел бы за пределы всей конструкции, пропустив `else`.

**Ожидаемый вывод**

```text
Символ отсутствует
```

````python
# Сохраняем строку, среди символов которой будем искать совпадение.
course_code = "PYTHON"
# Задаём букву A. В строке PYTHON этого символа нет.
search_symbol = "A"

# Последовательно проверяем каждый символ исходной строки, пока не найдём цель или не
# исчерпаем элементы.
for symbol in course_code:
    # Сравниваем текущий символ с буквой A. Все шесть сравнений в этом примере ложны.
    if symbol == search_symbol:
        # Эта строка сообщает о совпадении при выполнении условия.
        print("Символ найден")
        # Прерываем поиск при совпадении. Такое завершение пропустило бы else цикла.
        break
# Входим в ветку после нормального исчерпания элементов без break.
else:
    # Один раз сообщаем, что поиск завершился без совпадения.
    print("Символ отсутствует")
````

**Пример 95. Пропуск else цикла при найденном символе**

Цикл ищет букву `T` в строке `PYTHON`. Первый символ `P` и второй `Y` дают ложные сравнения. Третьим выделяется `T` на позиции `2`, поэтому `symbol == search_symbol` становится истинным.

При совпадении ячейка печатает `Символ найден`, затем выполняет `break`. Поиск заканчивается сразу, так что `H`, `O` и `N` больше не сравниваются с целью. Ветку `else`, которая расположена на уровне `for`, Python тоже пропускает из-за `break`.

Напечатать сообщение в ветке `if` недостаточно для подавления `else` цикла. Если убрать `break`, обход после буквы `T` продолжится до конца и появятся оба сообщения. Сравнение строк учитывает регистр. Если целью станет строчная `t`, совпадений в исходной строке не будет и отработает `else`.

**Ожидаемый вывод**

```text
Символ найден
```

````python
# Задаём строку из шести заглавных букв для поиска.
course_code = "PYTHON"
# Выбираем букву T, которая находится на третьей позиции строки.
search_symbol = "T"

# Читаем по одному символу слева направо, начиная с P.
for symbol in course_code:
    # Сравниваем очередной символ с T. Проверка истинна на третьем проходе.
    if symbol == search_symbol:
        # При совпадении сразу сообщаем, что символ найден.
        print("Символ найден")
        # Завершаем цикл после находки. Команда одновременно пропускает его ветку else.
        break
# Ветка принадлежит циклу for и выполнялась бы после исчерпания его элементов без break.
else:
    # Сообщение об отсутствии здесь не появится, поскольку поиск остановится на T.
    print("Символ отсутствует")
````

**Пример 96. Выполнение else после ложного условия while**

Счётчик `attempt` начинается с `1`. Перед каждым проходом `while` сравнивает его с верхней границей `3`. Когда условие истинно, ячейка выводит номер текущей попытки и увеличивает счётчик на единицу.

Проверки допускают номера `1`, `2` и `3`, поэтому появляются ровно три строки с попытками. После третьей прибавки `attempt` становится равным `4`. Сравнение `4 <= 3` даёт `False`, и Python переходит в `else`, где печатается `Все попытки использованы`.

У `else` такой же отступ, как у `while`. Он выполняется при завершении по ложному условию, даже если условие было ложным уже при первом входе и тело ни разу не запустилось. Если добавить в тело `break` для случая успешной попытки и выполнить его, ветка `else` будет пропущена. В этом примере `break` отсутствует, поэтому итоговое сообщение гарантированно появляется в конце обычного выполнения.

**Ожидаемый вывод**

```text
Попытка №1
Попытка №2
Попытка №3
Все попытки использованы
```

````python
# Инициализируем номер первой попытки значением 1.
attempt = 1

# Проверяем перед проходом, не превышает ли номер попытки 3.
while attempt <= 3:
    # Печатаем номер разрешённой текущей попытки до прибавки.
    print(f"Попытка №{attempt}")
    # Увеличиваем счётчик. После номера 3 значение 4 сделает условие цикла ложным.
    attempt += 1
# Переходим в ветку while после того, как условие стало ложным и цикл завершился без
# break.
else:
    # Печатаем итоговое сообщение один раз после всех трёх попыток.
    print("Все попытки использованы")
````

## 4.1 Список(list) как изменяемая последовательность;

**Пример 97. Создание списков и типы элементов**

Литерал списка записывается в квадратных скобках. Элементы перечисляются через запятую и сохраняют порядок. В `students` лежат три строки с именами, в `scores` — три целые оценки. **Пример 98. Индексы, срезы, длина списка и перебор**

В списке из пяти оценок индексы слева идут от `0` до `4`, а справа — от `-1` до `-5`. Поэтому `scores[0]` даёт первую оценку `85`, а `scores[-1]` — последнюю `95`. Индексация выбирает отдельный объект из списка. Например, обращение к позиции `5` вызвало бы `IndexError`.

Срез `scores[1:4]` берёт индексы `1`, `2` и `3`. Верхняя граница `4` исключается, поэтому получается `[90, 75, 60]`. В срезе `scores[::2]` начало и конец опущены, а шаг равен двум. Из позиций `0`, `2` и `4` получается `[85, 75, 95]`. Оба среза создают новые списки.

Вызов `len(scores)` возвращает количество элементов исходного списка, то есть `5`. Цикл `for score in scores` получает сами оценки по порядку, начиная с `85`, и выводит каждую на отдельной строке. Индекс нужен отдельно только тогда, когда программе важна позиция. Для перебора с позициями подходит `enumerate(scores)`.

**Ожидаемый вывод**

```text
85
95
[90, 75, 60]
[85, 75, 95]
5
85
90
75
60
95
```

````python
# Создаём исходный список из пяти оценок для всех последующих операций.
scores = [85, 90, 75, 60, 95]

# Читаем элемент с индексом 0. Это первая оценка 85.
print(scores[0])
# Отсчитываем один элемент справа по индексу -1. Получаем последнюю оценку 95.
print(scores[-1])
# Создаём срез с оценками на позициях 1, 2 и 3. Позиция 4 уже исключена.
print(scores[1:4])
# Создаём срез с шагом 2. Выбираем оценки на позициях 0, 2 и 4.
print(scores[::2])
# Печатаем число элементов исходного списка. Срезы его не изменили.
print(len(scores))

# На каждом проходе берём очередную оценку из исходного списка.
for score in scores:
    # Печатаем текущую оценку на отдельной строке.
    print(score)
````

**Пример 99. Замена элементов списка на месте**

Перед изменением программа печатает исходный список `[85, 90, 75, 60, 95]` и его идентификатор через `id(scores)`. Идентификатор относится к объекту списка и остаётся постоянным в течение его жизни.

Присваивание `scores[0] = 100` заменяет значение в первом элементе. Присваивание `scores[-1] = 80` заменяет последнее. Три средние оценки сохраняются на своих позициях. Получается `[100, 90, 75, 60, 80]`, длина по-прежнему равна пяти.

Оба вызова `id(scores)` показывают один идентификатор, поскольку изменилось содержимое существующего списка. В образце ниже `ID_СПИСКА` обозначает повторяющееся число. Присваивание по одному индексу требует существующей позиции. Например, `scores[5] = 50` для списка длины пять вызвало бы `IndexError`.

**Пример вывода с условным идентификатором**

```text
[85, 90, 75, 60, 95]
ID_СПИСКА
[100, 90, 75, 60, 80]
ID_СПИСКА
```

````python
# Создаём список из пяти оценок и связываем его с именем scores.
scores = [85, 90, 75, 60, 95]

# Печатаем содержимое списка перед заменой элементов.
print(scores)
# Запоминаем для сравнения идентификатор исходного объекта в выводе.
print(id(scores))

# Заменяем первую оценку 85 на 100, сохраняя позицию 0.
scores[0] = 100
# Заменяем последнюю оценку 95 на 80 по отрицательному индексу.
scores[-1] = 80

# Печатаем новое содержимое того же списка. Изменились только крайние элементы.
print(scores)
# Повторно печатаем идентификатор. Он совпадает с идентификатором до изменений.
print(id(scores))
````

Присваивание обычному срезу может увеличить, уменьшить или сохранить длину списка. Для среза с шагом, отличным от `1`, действуют другие правила. Например, при присваивании в `scores[::2]` количество новых элементов должно совпадать с количеством выбранных позиций.

**Пример 101. Два имени для одного изменяемого списка**

Присваивание `current_scores = scores` получает объект, уже связанный с именем `scores`, и связывает с ним второе имя. В памяти существует один список оценок. По любому из имён можно обратиться к его элементам.

Присваивание `current_scores[0] = 100` меняет первый элемент общего списка. При чтении через `scores` и через `current_scores` видно одно и то же содержимое `[100, 90, 75]`. Проверка `scores is current_scores` возвращает `True`, поскольку проверяет идентичность объектов.

Этот эффект особенно важен, когда программа сохраняет исходные данные и одновременно редактирует рабочую версию. Простое присваивание второго имени связывает его с тем же объектом. Для отдельного верхнего уровня списка можно создать копию с помощью `scores[:]` или `scores.copy()`.

**Ожидаемый вывод**

```text
[100, 90, 75]
[100, 90, 75]
True
```

````python
# Создаём список с тремя исходными оценками.
scores = [85, 90, 75]
# Связываем имя current_scores с тем же объектом. У списка теперь два имени.
current_scores = scores

# Меняем оценку по индексу 0 через имя current_scores. Меняется общий список.
current_scores[0] = 100

# Печатаем содержимое через имя scores и видим уже изменённую первую оценку.
print(scores)
# Печатаем то же содержимое через имя current_scores.
print(current_scores)

# Проверяем идентичность объектов оператором is. Оба имени ведут к одному списку, поэтому
# выводится True.
print(scores is current_scores)
````

**Пример 102. Независимый список через полный срез**

В `scores[:]` обе границы опущены, поэтому срез берёт все элементы исходного списка. Для обычного `list` он создаёт новый объект списка с теми же элементами в том же порядке. Сразу после присваивания оба списка равны по содержимому, но существуют отдельно.

Выражение `current_scores[0] = 100` заменяет ссылку в первом элементе новой копии. Исходный список сохраняет `85` в этой позиции. Первая печать показывает `[85, 90, 75]`, вторая — `[100, 90, 75]`. Оператор `is` возвращает `False`, потому что это два объекта списка.

Полный срез создаёт поверхностную копию. Он копирует список верхнего уровня, а содержащиеся в нём объекты остаются общими. Здесь элементы — неизменяемые целые числа, и операция заменяет позицию копии. При вложенных списках изменение общего внутреннего списка было бы видно из обеих внешних версий.

**Ожидаемый вывод**

```text
[85, 90, 75]
[100, 90, 75]
False
```

````python
# Создаём исходный список, содержимое которого хотим сохранить.
scores = [85, 90, 75]
# Копируем все элементы полным срезом. В current_scores находится новый список.
current_scores = scores[:]

# Заменяем первый элемент только в копии, сохраняя исходную позицию со значением 85.
current_scores[0] = 100

# Печатаем исходные оценки. Первая по-прежнему равна 85.
print(scores)
# Печатаем рабочую копию. На первой позиции теперь 100.
print(current_scores)

# Проверяем идентичность списков через is. Это два разных объекта, поэтому получаем False.
print(scores is current_scores)
````

**Пример 103. Добавление элементов через append, extend и insert**

Вначале в `students` два имени. Вызов `append("Александр")` добавляет один объект в конец существующего списка. Даже если аргумент содержит много символов, всё имя остаётся одним элементом. После этого в списке три имени.

Метод `extend()` перебирает переданный объект и добавляет каждый полученный элемент отдельно. В этом примере он берёт сначала `"Анна"`, затем `"Дмитрий"`, доводя длину списка до пяти. Если передать тот же список в `append()`, он окажется одним вложенным элементом. Если передать строку в `extend()`, она будет перебрана посимвольно.

В `insert(1, "Ольга")` индекс `1` означает вторую позицию. Новое имя появляется перед `"Мария"`, а Мария и следующие за ней имена сдвигаются вправо. Все три метода меняют исходный список на месте и возвращают `None`. Именно поэтому после каждого вызова печатается `students`.

**Ожидаемый вывод**

```text
['Иван', 'Мария', 'Александр']
['Иван', 'Мария', 'Александр', 'Анна', 'Дмитрий']
['Иван', 'Ольга', 'Мария', 'Александр', 'Анна', 'Дмитрий']
```

````python
# Создаём список из двух имён. Иван занимает позицию 0, Мария позицию 1.
students = ["Иван", "Мария"]

# Добавляем одно имя в конец. Александр становится третьим элементом.
students.append("Александр")
# Печатаем три имени в порядке добавления.
print(students)

# Расширяем список двумя элементами переданного списка. Анна и Дмитрий добавляются по
# отдельности.
students.extend(["Анна", "Дмитрий"])
# Печатаем пять имён после расширения.
print(students)

# Вставляем Ольгу перед элементом с индексом 1. Мария и все следующие элементы сдвигаются
# вправо.
students.insert(1, "Ольга")
# Печатаем итоговую последовательность из шести имён.
print(students)
````

**Пример 104. Удаление по значению, извлечение по индексу и очистка**

Метод `remove("Мария")` находит первый элемент, равный строке `"Мария"`, и удаляет его. В списке остаются Иван, Анна и Дмитрий в прежнем относительном порядке. Если совпадающих элементов несколько, один вызов удалит только первый. Если совпадений нет, метод вызовет `ValueError`.

Вызов `pop(0)` обращается к позиции `0`, удаляет оттуда Ивана и возвращает это имя. Возвращённая строка сохраняется в `removed_student`, а список сокращается до Анны и Дмитрия. Без аргумента `pop()` извлекает последний элемент. Обращение к отсутствующему индексу или вызов для пустого списка приводит к `IndexError`.

Вызов `clear()` удаляет все оставшиеся элементы. Сам объект списка сохраняется. Если на него ссылалось другое имя, через это имя тоже будет доступен пустой список. Методы `remove()` и `clear()` возвращают `None`, а `pop()` возвращает удалённый объект.

**Ожидаемый вывод**

```text
['Иван', 'Анна', 'Дмитрий']
Иван
['Анна', 'Дмитрий']
[]
```

````python
# Создаём список с четырьмя именами для трёх последовательных операций удаления.
students = ["Иван", "Мария", "Анна", "Дмитрий"]

# Удаляем первое вхождение имени Мария по значению. Индекс передавать не нужно.
students.remove("Мария")
# Показываем список после удаления Марии. Иван, Анна и Дмитрий остаются.
print(students)

# Удаляем элемент с индексом 0 и сохраняем возвращённое имя Иван.
removed_student = students.pop(0)
# Печатаем строку, которую вернул метод pop.
print(removed_student)
# Печатаем список после извлечения. В нём остались Анна и Дмитрий.
print(students)

# Удаляем все элементы существующего списка, сохраняя сам объект.
students.clear()
# Печатаем пустой список в виде квадратных скобок.
print(students)
````

**Пример 105. Поиск позиции и подсчёт вхождений**

Метод `index(90)` просматривает список слева направо и возвращает индекс первого совпадения. Первый элемент `85` находится на позиции `0`, а следующая за ним оценка `90` — на позиции `1`. Поэтому первый вызов выводит `1`.

Второй аргумент в `index(90, 2)` задаёт начальный индекс поиска. Проверка начинается с позиции `2`, где лежит `75`, и находит следующую оценку `90` на позиции `3`. Возвращается индекс относительно всего списка. Можно указать и третий аргумент `stop`, который исключает позиции начиная с указанной верхней границы.

Метод `count()` возвращает количество совпадающих элементов во всём списке. Оценка `90` встречается на позициях `1`, `3` и `5`, поэтому результат равен трём. Оценки `100` нет, поэтому `count(100)` возвращает ноль. Для той же отсутствующей оценки `index(100)` вызвал бы `ValueError`. Оба метода оставляют список в исходном порядке.

**Ожидаемый вывод**

```text
1
3
3
0
```

````python
# Создаём список из шести оценок. Оценка 90 встречается на трёх разных позициях.
scores = [85, 90, 75, 90, 60, 90]

# Ищем первое вхождение 90 с начала списка. Находим абсолютный индекс 1.
print(scores.index(90))
# Начинаем поиск с индекса 2 и получаем индекс 3. Первое вхождение на позиции 1 пропущено.
print(scores.index(90, 2))
# Считаем все три вхождения оценки 90. Поиск продолжается до конца списка.
print(scores.count(90))
# Ищем количество сотен. При отсутствии совпадений count возвращает 0.
print(scores.count(100))
````

**Пример 106. Сортировка по возрастанию и убыванию и разворот списка**

Метод `sort()` переставляет элементы в самом списке. По умолчанию числа упорядочиваются по возрастанию, поэтому после первого вызова получается `[60, 75, 85, 90, 95]`. Метод возвращает `None`. Чтобы получить отдельный отсортированный список и сохранить исходный порядок, подходит `sorted(scores)`.

Параметр `reverse=True` меняет направление сортировки. После второго вызова сначала идёт наибольший балл `95`, а в конце наименьший `60`. В обоих вызовах `sort()` порядок определяется сравнением числовых значений.

Метод `reverse()` разворачивает текущую последовательность без сравнений. Здесь он превращает убывающий ряд в возрастающий, поскольку вход уже отсортирован по убыванию. Для произвольного порядка результат будет просто обратным. Например, разворот `[3, 1, 2]` даёт `[2, 1, 3]`. Метод меняет список на месте и возвращает `None`.

**Ожидаемый вывод**

```text
[60, 75, 85, 90, 95]
[95, 90, 85, 75, 60]
[60, 75, 85, 90, 95]
```

````python
# Создаём список оценок в исходном произвольном порядке.
scores = [85, 90, 75, 60, 95]

# Сортируем тот же список по возрастанию чисел.
scores.sort()
# Показываем возрастающую последовательность от 60 до 95.
print(scores)

# Снова сортируем тот же список и включаем порядок по убыванию.
scores.sort(reverse=True)
# Показываем убывающую последовательность от 95 до 60.
print(scores)

# Разворачиваем текущий порядок элементов на месте. Сравнивать числа для этого не нужно.
scores.reverse()
# Печатаем итоговый порядок. Он снова возрастающий, поскольку развёрнут убывающий ряд.
print(scores)
````

**Пример 107. Поверхностная копия списка через copy**

Метод `scores.copy()` создаёт отдельный список верхнего уровня с теми же элементами в том же порядке. Для обычного списка он даёт тот же результат, что и `scores[:]`. Сразу после копирования списки равны по содержимому, но уже имеют разную идентичность.

Вызов `new_scores.append(100)` добавляет оценку только в новый список. Исходный список сохраняет три элемента `[85, 90, 75]`, а копия содержит четыре `[85, 90, 75, 100]`. Оператор `is` проверяет идентичность и возвращает `False`, потому что имена связаны с двумя отдельными объектами.

Копия поверхностная, поскольку элементы копируются как ссылки на исходные объекты. Для целых чисел это позволяет независимо добавлять или заменять позиции в копии. Если же элементом является вложенный изменяемый список, изменение его содержимого будет видно через обе версии. Для копирования вложенной структуры со своими элементами можно рассмотреть `copy.deepcopy()`.

**Ожидаемый вывод**

```text
[85, 90, 75]
[85, 90, 75, 100]
False
```

````python
# Создаём исходный список из трёх оценок.
scores = [85, 90, 75]

# Создаём отдельный список с теми же тремя элементами.
new_scores = scores.copy()
# Добавляем сотню только в новый список. У исходного списка длина остаётся равной трём.
new_scores.append(100)

# Печатаем исходный список без добавленного элемента.
print(scores)
# Печатаем копию с дополнительной оценкой 100 в конце.
print(new_scores)

# Проверяем идентичность объектов. Два разных списка дают False.
print(scores is new_scores)
````

**Пример 108. Два уровня индексов и длина вложенных списков**

Внешний список `scores` содержит три элемента. Каждый элемент сам является списком оценок одного студента. Выражение `scores[0]` выбирает первый вложенный список целиком. В выводе его три оценки видны вместе с квадратными скобками.

Две пары скобок читаются слева направо. В `scores[1][2]` сначала выбирается второй список `[60, 80, 95]`, затем его третий элемент `95`. Аналогично `scores[2][0]` берёт первую оценку третьего студента и даёт `100`. На каждом уровне отсчёт начинается с нуля.

Вызов `len(scores)` считает вложенные списки, поэтому возвращает число студентов в этом примере. Вызов `len(scores[0])` считает оценки только первого студента. Оба результата здесь равны трём, потому что совпало число студентов и число оценок в первой строке. Вложенные списки могут иметь разные длины, поэтому перед индексированием конкретной строки важно учитывать именно её длину.

**Ожидаемый вывод**

```text
[85, 90, 75]
95
100
3
3
```

````python
# Открываем внешний список. Его элементами будут отдельные списки оценок.
scores = [
    # Создаём первую строку с тремя оценками. Во внешнем списке она получит индекс 0.
    [85, 90, 75],
    # Создаём вторую строку с индексом 1. Её последняя оценка 95 находится на внутренней
    # позиции 2.
    [60, 80, 95],
    # Создаём третью строку с индексом 2. Её первая оценка 100 находится на внутренней
    # позиции 0.
    [100, 90, 85]
# Закрываем внешний список из трёх вложенных списков.
]

# Печатаем весь первый вложенный список с оценками 85, 90 и 75.
print(scores[0])
# Выбираем второй список по индексу 1 и печатаем его третий элемент с индексом 2.
print(scores[1][2])
# Выбираем третий список и печатаем его первый элемент.
print(scores[2][0])

# Считаем элементы внешнего списка. Каждая строка учитывается как один элемент.
print(len(scores))
# Считаем оценки только в первой строке.
print(len(scores[0]))
````

**Пример 109. Изменение оценки, расширение строки и добавление строки**

Присваивание `scores[0][1] = 100` выбирает первый вложенный список и заменяет только его вторую оценку. Было `[85, 90, 75]`, после замены получается `[85, 100, 75]`. Число оценок у первого студента сохраняется.

В вызове `scores[1].append(70)` получателем метода является второй вложенный список. В него добавляется ещё одна оценка, и он превращается в `[60, 80, 95, 70]`. Внешний список на этом шаге по-прежнему содержит три строки.

Последний вызов `scores.append([80, 85, 90])` обращается к внешнему списку. В него добавляется целиком новый вложенный список, поэтому строк становится четыре. Длины строк теперь равны `3`, `4`, `3` и `3`. У списков Python нет требования, чтобы все вложенные строки были одной длины.

**Ожидаемый вывод**

```text
[[85, 100, 75], [60, 80, 95, 70], [100, 90, 85], [80, 85, 90]]
```

````python
# Открываем внешний список для набора оценок нескольких студентов.
scores = [
    # Задаём первую строку. Её элемент 90 с индексом 1 будет заменён.
    [85, 90, 75],
    # Задаём вторую строку, в которую затем добавим четвёртую оценку.
    [60, 80, 95],
    # Задаём третью строку. Последующие операции сохранят её содержимое.
    [100, 90, 85]
# Закрываем внешний список. Первоначально в нём три строки.
]

# Находим первый список и меняем его вторую оценку с 90 на 100.
scores[0][1] = 100
# Добавляем число 70 в конец второго вложенного списка. Меняется только длина этой строки.
scores[1].append(70)
# Добавляем целый список оценок как четвёртый элемент внешнего списка.
scores.append([80, 85, 90])

# Печатаем результат всех трёх изменений с сохранением вложенных скобок.
print(scores)
````

**Пример 110. Средний балл каждого студента во вложенных циклах**

На каждом проходе внешний цикл получает один вложенный список. Перед обработкой очередного студента переменная `total_score` снова получает `0`. Внутренний цикл последовательно прибавляет к ней все оценки выбранного студента. Если вынести обнуление за пределы внешнего цикла, суммы разных студентов начнут накапливаться вместе.

В первом списке накопленная сумма проходит значения `85`, `175` и `250`. Деление `250 / 3` даёт примерно `83.3333`. Во втором списке сумма равна `235`, а в третьем `275`. Для них средние равны примерно `78.3333` и `91.6667`. В каждом случае делитель берётся из `len(student_scores)`, то есть из длины текущей строки.

Формат `.2f` показывает ровно две цифры после точки с округлением для вывода. Поэтому последняя строка заканчивается значением `91.67`. Пустой вложенный список потребовал бы отдельной проверки, поскольку его длина равна нулю и формула вызвала бы `ZeroDivisionError`.

**Ожидаемый вывод**

```text
Средний балл: 83.33
Средний балл: 78.33
Средний балл: 91.67
```

````python
# Открываем внешний список, в котором каждый элемент соответствует одному студенту.
scores = [
    # Задаём оценки первого студента. Их сумма равна 250.
    [85, 90, 75],
    # Задаём оценки второго студента. Их сумма равна 235.
    [60, 80, 95],
    # Задаём оценки третьего студента. Их сумма равна 275.
    [100, 90, 85]
# Закрываем набор из трёх списков оценок.
]

# Получаем очередной вложенный список целиком и обрабатываем его как одного студента.
for student_scores in scores:
    # Обнуляем накопитель внутри внешнего цикла, чтобы сумма начиналась заново для каждой
    # строки.
    total_score = 0

    # Перебираем отдельные оценки текущего студента во внутреннем цикле.
    for score in student_scores:
        # Прибавляем текущий балл к сумме оценок выбранного студента.
        total_score += score

    # После внутреннего цикла делим накопленную сумму на число оценок в текущем списке.
    average_score = total_score / len(student_scores)

    # Печатаем средний балл текущего студента с двумя цифрами после точки.
    print(f"Средний балл: {average_score:.2f}")
````

**Пример 111. Общие вложенные списки после поверхностного копирования**

Метод `scores.copy()` создаёт новый внешний список. В его элементы попадают ссылки на те же две строки, которые уже лежат в `scores`. Поэтому сравнение `scores is new_scores` даёт `False`, а `scores[0] is new_scores[0]` даёт `True`.

При выполнении `new_scores[0][0] = 100` первая пара скобок выбирает общий вложенный список, а вторая указывает его первый элемент. Значение `85` в общей строке заменяется на `100`. После этого оба внешних списка при печати показывают одинаковую вложенную структуру `[[100, 90], [75, 80]]`.

Уровень изменения определяет, где оно будет заметно. Например, если присвоить `new_scores[0] = [100, 90]`, новое значение займёт позицию только в копии внешнего списка. В текущем примере доступ продолжается глубже через `[0][0]`, поэтому изменяется содержимое общей строки.

**Ожидаемый вывод**

```text
[[100, 90], [75, 80]]
[[100, 90], [75, 80]]
False
True
```

````python
# Создаём внешний список с двумя отдельными вложенными строками.
scores = [[85, 90], [75, 80]]

# Копируем внешний список. Обе вложенные строки пока общие для двух списков.
new_scores = scores.copy()
# Выбираем общую первую строку через копию и заменяем в ней значение 85 на 100.
new_scores[0][0] = 100

# Печатаем исходный внешний список. Изменение видно в его общей первой строке.
print(scores)
# Печатаем копию. Она обращается к той же изменённой строке.
print(new_scores)

# Сравниваем идентичность внешних списков. Метод copy создал отдельный объект, поэтому
# получаем False.
print(scores is new_scores)
# Сравниваем первые вложенные списки. Это один общий объект, поэтому получаем True.
print(scores[0] is new_scores[0])
````

**Пример 112. Глубокое копирование вложенных списков**

Модуль `copy` входит в стандартную библиотеку Python. Функция `copy.deepcopy(scores)` для такой структуры создаёт отдельный внешний список и копирует каждую из двух вложенных строк. В момент копирования значения в исходной структуре и копии совпадают.

При присваивании `new_scores[0][0] = 100` выбирается первая строка копии и меняется её первый элемент. Первая строка оригинала продолжает начинаться с `85`. Две команды печати позволяют проверить это по содержимому, а выражение `scores[0] is new_scores[0]` даёт `False` и подтверждает, что строки являются разными объектами.

Копирование проходит по структуре исходных объектов. Если в исходном списке одна и та же строка встречается в нескольких позициях, `deepcopy()` сохранит это совместное использование внутри копии. В этом примере обе исходные строки созданы отдельными литералами, и в копии им соответствуют две отдельные строки.

**Ожидаемый вывод**

```text
[[85, 90], [75, 80]]
[[100, 90], [75, 80]]
False
```

````python
# Импортируем модуль copy, чтобы обратиться к функции deepcopy.
import copy

# Создаём исходный список с двумя строками оценок. У каждой строки свой объект.
scores = [[85, 90], [75, 80]]

# Глубоко копируем список. В копии появляются отдельный внешний список и отдельные
# вложенные строки.
new_scores = copy.deepcopy(scores)
# Заменяем первую оценку только в первой строке копии.
new_scores[0][0] = 100

# Печатаем исходные оценки. Первая строка сохраняет число 85.
print(scores)
# Печатаем копию. Её первая строка теперь начинается с числа 100.
print(new_scores)

# Проверяем идентичность первых строк. Это разные списки, поэтому выводится False.
print(scores[0] is new_scores[0])
````

**Пример 113. Почему умножение списка создаёт общие строки**

Во внутреннем литерале `[0, 0, 0]` создаётся одна строка. Внешний литерал `[[0, 0, 0]]` содержит ссылку на эту строку. Оператор `* 3` повторяет элементы внешнего списка и получает три позиции, каждая из которых ссылается на одну и ту же строку.

Выражение `matrix[0][0] = 1` выбирает строку в позиции `0` и заменяет её первый элемент. Поскольку остальные позиции внешнего списка содержат ту же строку, при печати значение `1` видно в начале каждой из трёх строк. Здесь изменился первый элемент одного общего объекта.

Связь можно увидеть по проверке `matrix[0] is matrix[1]`, которая даст `True`. Такая конструкция подходит, если повторное использование одного объекта задумано. Для матрицы с независимо изменяемыми строками нужно создать новый вложенный список для каждой позиции, как в следующем примере.

**Ожидаемый вывод**

```text
[[1, 0, 0], [1, 0, 0], [1, 0, 0]]
```

````python
# Создаём одну строку из трёх нулей и трижды повторяем ссылку на неё во внешнем списке.
matrix = [[0, 0, 0]] * 3

# Меняем первый элемент общей строки через первую позицию внешнего списка.
matrix[0][0] = 1

# Печатаем матрицу. Каждая позиция показывает одну и ту же изменённую строку.
print(matrix)
````

**Пример 114. Независимые строки матрицы через цикл**

Пустой список `matrix` будет внешним контейнером. Цикл по `range(3)` выполняет тело три раза. На каждом проходе выражение `[0, 0, 0]` заново создаёт отдельный список и передаёт его методу `append()`. Значение переменной `row` в теле не используется. Диапазон задаёт количество строк.

После трёх проходов значения всех строк одинаковы, но сами строки являются разными объектами. Поэтому присваивание `matrix[0][0] = 1` меняет только первый элемент первой строки. Вторая и третья строки сохраняют по три нуля.

Проверка `matrix[0] == matrix[1]` до изменения дала бы `True`, поскольку содержимое совпадает. Проверка `matrix[0] is matrix[1]` уже в тот момент дала бы `False`. Если вынести создание строки до цикла и добавлять её по одному имени на каждом проходе, все позиции снова будут ссылаться на один объект.

**Ожидаемый вывод**

```text
[[1, 0, 0], [0, 0, 0], [0, 0, 0]]
```

````python
# Создаём пустой внешний список. Строки матрицы появятся в цикле.
matrix = []

# Выполняем тело три раза. Переменная row последовательно принимает 0, 1 и 2.
for row in range(3):
    # На каждом проходе создаём новый список из трёх нулей и добавляем его как одну
    # строку.
    matrix.append([0, 0, 0])

# Меняем только первую ячейку первой строки. Остальные строки являются отдельными
# объектами.
matrix[0][0] = 1

# Печатаем матрицу. Единица видна только в левом верхнем углу.
print(matrix)
````
